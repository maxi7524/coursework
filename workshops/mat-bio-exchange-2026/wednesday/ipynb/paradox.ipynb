## Model: resource–consumer dynamics

The model constructed below is:

$$\begin{aligned}
\frac{dA}{dt} &= rA\left(1 - \frac{A}{k}\right) - \frac{cAZ}{s+A},\\
\frac{dZ}{dt} &= -qZ + \frac{pcAZ}{s+A}.
\end{aligned}$$

Here $A$ is the resource/prey and $Z$ is the consumer/predator. The first term is logistic growth of $A$ (intrinsic rate $r$, carrying capacity $k$). The fraction $AZ/(s+A)$ is a saturating (Holling type II) consumption term: $c$ is the maximum interaction rate and $s$ controls saturation. Consumers die at rate $q$ and reproduce from consumed resource with conversion efficiency $p$.

The parameter vector `p` below supplies these values to the model; explicitly passing it to each simulation makes clear which parameter set produced each plot.


read in grindR


```r
remotes::install_git("https://bitbucket.org/stan_maree/grind.git")
library(Grind)

```

first the model is defined


```r
model <- function(t, state, parms) {
  with(as.list(c(state,parms)), {
    
    dA <- r*A*(1 - A/k) - c*A*Z/(s+A) + 
    dZ <- - q*Z + p*c*A*Z/(s+A)
    
    return(list(c(dA, dZ)))  
  }) 
}  

```

then the parameter values are given


```r
p <- c(r=4,k=50,c=10,s=10,q=0.75,p=0.1)  # p is a named vector of parameters

```

and finally the begin state of the variables is given


```r
s <- c(A=1,Z=1)                          # s is the state, the initial values of the variables

```

run a simulation for tmax timesteps, generate an output datapoint every tstep


```r
run(tmax=200, tstep=0.1, parms=p)

```

now first vary k and then each time make a new timeplot
(e.g. increase k in steps of 5 from 20 to 100)


```r
run(tmax=200, tstep=0.1,parms=c(r=4,k=20,c=10,s=10,q=0.75,p=0.1))

```

```r
plane(xmax=120,ymax=15,vector=TRUE)
run(tmax=200, tstep=0.1, parms=p, traject=TRUE)

newton(jacobian=T,plot=T,state=c(A=0,Z=0))
newton(jacobian=T,plot=T,state=c(A=10,Z=0))
newton(jacobian=T,plot=T,state=c(A=20,Z=30))
```
