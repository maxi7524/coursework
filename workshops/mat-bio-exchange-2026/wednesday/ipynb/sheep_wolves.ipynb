## Model: source–attacker dynamics

Here $S$ denotes the source (sheep) and $A$ the attacker (wolf). The equations are:

$$\begin{aligned}
\frac{dS}{dt} &= g_S S - d_S S - aAS,\\
\frac{dA}{dt} &= g_A aAS - d_A A.
\end{aligned}$$

The source grows at rate $g_S$, dies at rate $d_S$, and is removed through encounters with attackers at rate $a$. The coefficient $g_A$ converts consumed source into attacker growth; attackers die at rate $d_A$.


read in grindR


```r
remotes::install_git("https://bitbucket.org/stan_maree/grind.git")
library(Grind)

```

first the model is defined
dS/dt captures the source dynamics, note that we just type dS
dA/dt captures the attacker dynamics, note that we just type dW



```r
model <- function(t, state, parms) {
  with(as.list(c(state,parms)), {

    #this model describes wolves (A) eating sheep (S)   
    dS <- g_S*S-d_S*S-a*A*S
    dA <- g_A*a*A*S-d_A*A
    
#make sure to return all temporal derivatives      
    return(list(c(dS,dA)))  
  }) 
}

```

then the parameter values are given
`g_S` is the birth rate of source, for example 1.5/year (i.e. 3 lambs per ewe per year)
`d_S` is the natural death rate of source, for example 0.14/year (in the wild, source can 
become around 7 years old)
`a` is the capturing rate of source by attackers. This parameter very much depends on the terrain
and the exact units in which the variables are expressed (number of source or
concentration of source per what kind of area). Here we put it to 1 per attacker per year.
`g_A` is the amount of attacker offspring per captured source. If source and attackers are 
expressed in biomass, the value is expected to be close to 0.1 attacker per source,
using a typical 10% conversion efficiency when moving up one trophic level. 
If the variables are instead expressed in number or density of source and attackers, 
we also have to take the typical weight of attackers and source into account. While
attackers weigh around 35 kg, source weigh around 70 kg. When thinking in numbers or density,
the value of g_A should therefore rather be 0.2 attacker/source
`d_A` is the natural death rate of attackers. in the wild, attackers become around 8 year old, 
a reasonable value for d_A is thus 0.125/year.



```r
p <- c(g_S=1.5,d_S=0.14,a=1,g_A=0.2,d_A=0.125) # p is a named vector of parameters
```

and finally the initial conditions (begin state) of the variables is given


```r
s <- c(S=1,A=1) # s is the state, the initial values of the variables

```

run a simulation for tmax timesteps, generate an output datapoint every tstep
run the model for 100 years


```r
run(tmax=100, tstep=0.1, parms=p)
```

run the model for 100 years


```r
run(tmax=100, tstep=0.1, parms=p, xlab="Time [years]",ylab="Population size",)
```

start with no source


```r
s <- c(S=0,A=1)
```

run the model again


```r
run(tmax=100, tstep=0.1, parms=p)
```

without any food, the attackers present an exponential decay


```r
run(tmax=100,tstep=0.1,parms=p,show=c("A"),ymin=1e-5,log="y",xlab="Time [years]",ylab="Attacker population")

```

the most relevant arguments of run() that you might want to change one day, and their default values


```r
run(tmax=100, tstep=1, state=s, parms=p, odes=model, ymin=0, ymax=NULL,
  log="", xlab="Time", ylab="Density", tmin=0, draw=lines,
  show=NULL, timeplot=TRUE, traject=FALSE,
  table=FALSE, add=FALSE, legend=TRUE, solution=FALSE, lwd=2)

```

```r
plane(xmax=5,ymax=5,vector=TRUE)
s <- c(S=1,A=1)
run(tmax=100, tstep=0.1, parms=p,traject = TRUE)

```
