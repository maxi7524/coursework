```python
# znajdujemy to w GenBank
handle = Entrez.efetch(db='nuccore', id='AF216311', rettype='gb', retmode='text')
gb_record = SeqIO.read(handle, 'genbank')
#TODO jakie są rodzaje db=... i jak to konfigurować 
for feature in gb_record.features:
    if feature.type == "CDS":
        print(feature.qualifiers["protein_id"])

print(gb_record)
```

```python
# pobranie proteiny  ale tutaj potrzebujemy protein id
# UWAGA tutaj wykorzystuje informacje z poprzedniego 
protein_id = feature.qualifiers["protein_id"][0]

handle = Entrez.efetch(db="protein", id=protein_id, rettype="fasta", retmode="text")
protein_record = SeqIO.read(handle, "fasta")

print(protein_record)
```

```python
# wywołanie blast
import subprocess
# zapisujmey lokalnie nasze pliki 
SeqIO.write(protein_record, 'input.fasta', 'fasta')

# użycie lokalnych programów
cmd = [
    "blastp",
    "-query", "input.fasta",
    "-db", "nr",
    "-outfmt", "5",
    "-max_target_seqs", "500",
    "-out", "blast.xml"
]
subprocess.run(cmd)

# trzeba baze danych sobie tutaj zapisać (globalnie trzeba to zrobić)

# 
```

TO PIERWSZE 'ŁADOWANIW' plików 
Najważniejsze:
db

"nuccore" – dane nukleotydowe

"protein" – białka

"gene" – rekordy genów (nie sekwencje!)

"nucleotide" – alias dla nuccore

"assembly", "biosample", itd.

id

accession number (np. AF216311)

gi number (już nieużywany)

protein_id (np. NP_012345.1)

rettype

"fasta" – czysta sekwencja

"gb" – GenBank (cechy, CDS, annotacje)

"xml" – struktury XML

"gp" – GenPept

retmode

"text" – zwykły tekst

"xml" – XML

"asn.1" – format wewnętrzny NCBI
