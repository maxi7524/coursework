```python
# libraries
import numpy as np
import pandas as pd
from Bio import Entrez, SeqIO
import subprocess

# data
lista = [
    "VDFWAE",
    "VDFWAP",
    "VDFWAE",
    "VDFWAE",
    "VDFWAP",
    "VDFSAT",
    "VDFSAT",
    "VDFSAT",
    "VDFSAT",
    "VDFSAT",
    "VDFSAE",
    "LDFYAT",
    "LDFYAT",
    "VDFTAS",
    "ADFYAD"
]
```

```python
# zliczamy elementy, ilośc elementów na każdej pozycji 
aminoacids = 'ACDEFGHIKLMNPQRSTVWY'
# zliczamy ilość aminokwasów na każdej pozycji 
# (L, A) gdzie L - długość sekwencji, A - zbiór aminokwasów. 
M = np.zeros([len(lista[-1]), len(aminoacids)])
for i in range(len(lista[-1])):
    for j in range(len(aminoacids)):
        # zliczanie 
        value = len([True for s in lista if s[i] == aminoacids[j]])
        # przypisanie
        M[i, j] = value
print(M.sum(axis=1)) # ma sens 

M
```

```python
# accession 
accession_number = 'P49787' # to jest źle 
# downloading 
Entrez.email = 'maxstrozyk@gmail.com'
handle = Entrez.efetch(db="protein", id=accession_number, rettype="fasta", retmode="text")
# loading sequence
seq_record = SeqIO.read(handle, "fasta")
seq_record
```

```python
import os
# wywołanie blast

# zapisujmey lokalnie nasze pliki 
SeqIO.write(seq_record, 'input.fasta', 'fasta')

# użycie lokalnych programów UWAGA - musi być 
cmd = [
    "blastp",
    "-query", "input.fasta",
    "-db", os.path.expanduser("~/blastdb/swissprot/swissprot"),
    "-outfmt", "11",               # archiwum ASN.1
    "-max_target_seqs", "500",
    "-out", "zadanie2_blast.asn"
]
subprocess.run(cmd)

cmd = [
    "psiblast",
    "-query", "input.fasta",
    "-db", os.path.expanduser("~/blastdb/swissprot/swissprot"),
    "-num_iterations", "10",
    "-max_target_seqs", "500",
    "-outfmt", "11",
    "-out", "zadanie2_psi_blast.asn"
]
subprocess.run(cmd)
```

```python

subprocess.run([
    "blast_formatter",
    "-archive", "zadanie2_psi_blast.asn",
    "-html",
    "-out", "zadanie2_psi_blast.html"
])
subprocess.run([
    "blast_formatter",
    "-archive", "zadanie2_blast.asn",
    "-html",
    "-out", "zadanie2_blast.html"
])
```
