```python
%%bash
### AWK ###
file_allover='pd1.23/allover-blast-NAMv5-res4st'
# 0. -info - chcemy zobaczyć indeksy kolumn
# wszędzie stosuje head -3 w celu uzyskania czytelności 
head -2 $file_allover
# do wszystkich: - patrzymy w jaki sposób są te pliki rozdzielone, ustawiamy odpowiedni separator ',', print wyświetla cały wiersz domyślnie

# 1. widzimy, żę za odpowiednią wartość odpowiada indeks 4, 
echo -e '\nzadanie 1'
awk -F',' '$4 < 1e-18 {print}' $file_allover | head -3
# 2. stosuje 
echo -e '\nzadanie 2'
awk -F',' '$NF ~ /[cC]alcium/ {print}' $file_allover | head -3
# 3. widzimy że 7 i 8 kolumna odpowiada za wartości boolean, wybieramy kolumny w których jeden lub (||) drugi warunek jest spełniony
echo -e '\nzadanie 3' 
awk -F','   '$7==1 || $8==1 {print}' $file_allover | head -3
```

```python
%%bash
### GREP ###
file_allover='pd1.23/allover-blast-NAMv5-res4st'
file_protein='pd1.23/4577.protein.aliases.v12.04st.txt'
# do wszystkich - w grep'ie musimy stosować podstawowy regex -> trzeba oznaczać {} -> \{\},

# 4. stosujemy ekspresje regularne, korzystamy z [0-9] zamiast z [[:digit:]] bo jest krócej
echo 'zadanie 4'
grep 'EC [23]\.[0-9]\.[0-9]\.[0-9]\{2\}' $file_allover | head -3
# 5. tutaj musimy narzucić żeby grep znalazł wartości w odpowiedniej kolumnie, narzucamy żeby znajdywał odpowiednio początek i koniec, oraz znajdywał warunek wewnątrz. 
echo -e '\nzadanie 5'
head -2 $file_protein 
grep '^4577\..*[A-Z]\{2,\}[0-9]\{2,\}.* ' $file_protein | head -3
```

```python
%%bash
### SED ###
file_allover='pd1.23/allover-blast-NAMv5-res4st'
# 6. s -> zamiana znaku g-> globalnie (wszystkie wystąpienia)
echo 'zadanie 6'
sed 's/|/ /g' $file_allover | head -2
# 7. tutaj jest delikatność, musimy uwzględnić wszystkie nawiasy, nie możemy dopuścić do następującej sytuacji:
# (...)!!!(...) -> weźmie pod uwagę zewnętrzne nawiasy bo fragment !!! jest porządany, zatem ustalamy taki regex, żeby znajdywał "(" wyrzucał wszystkie znaki poza ")" i potem brał uwagę resztę wystąpień ")". Stosujemy potem g żeby brało pod uwagę wszystkie wystąpienia. 
echo -e '\nzadanie 7'
sed 's/([^)]*)//g' $file_allover | head -2
```

```python
%%bash
### SORT + JOIN ###
file_1='pd1.23/4577.kegg_id_mini'
file_2='pd1.23/dane.kegg_mini'
file_3='joined.txt'
# chce informacje o kolumnach
echo -e 'Zadanie 8\ninformacja 1'
head -2 $file_1
echo -e '\ninformacja 2'
head -2 $file_2

# sortujemy pliki - potrzebujemy odpowiedniej kolejności identyfikatorów,
sort -k1,1 $file_1 > ${file_1}.sorted
# stosujemy dodatkowo awk którym sortujemy odpowiednie kolumny - zachowuje tabulator jako odstęp
sort -k1,1 $file_2 | awk '{print $1, $2, $5, $6}' > ${file_2}.sorted 

# łączymy pliki - informuje o kolumnach widzieliśmy że są to pierwsze kolumny, że mieć kolejność w zadaniu z zamysłem wybieramy file_1 jako pierwszy do złączenia. 
join -1 1 -2 1 ${file_1}.sorted ${file_2}.sorted > $file_3

# końcowy wynik
echo -e '\nkońcowy wynik'
head -3 $file_3
echo -e '\nliczba wierszy'
wc -l $file_3
```
