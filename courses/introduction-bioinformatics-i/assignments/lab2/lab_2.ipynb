grep

Służy do wybierania wierszy z plików które zawierają odpowiedni wzorzec 

Ogólnie

Znaki specjale
- '' 
    - nie są wyszukiwane 
    - służą do pozbywania się błędów z powodu specjalnych znaków

    
linia po linii
- wszystkie działania są wykonywane dla każdej linii

```python
# zadanie 1
aa_path = 'pliki2/aa'

!grep 'Ala' {aa_path}
print('\n')
!grep 'C6H' {aa_path}
```

```python
# zadanie 2
fa = 'pliki2/mature.fa'

# wyszukiwanie sekwencji
# linia opis > poprzedzona tym znakiem -> będziemy szukać tego znaku i wyświetlać co jest po nim

# !grep '' {fa}
!grep '>.' {fa}
!grep '>[^]*p' 
```

```python
# zadanie 3
var = 'oryza sativa' 
fl = 'pliki2/mature.fa'

!grep -i '{var}' {fl} | head
print('\n Wariant drugi')
var = '[oO]ryza sativa' 
!grep '{var}' {fl} | head
```

```python
# zadanie 4
var = 'UUUUU'
fl = 'pliki2/mature.fa'

print('Kotwica poczatek')
!grep '^{var}' {fl} | head

print('Kotwica koniec')
!grep '{var}$' {fl} | head
```

```python
# zadanie 5
var = 'U\\{5\\}'
fl = 'pliki2/mature.fa'

print('Zmiana formatu')
!grep '^{var}' {fl} | head

for i in range(2, 4):
    print(f'Ograniczenie do {i}')
    var = 'U\\{' + f'{i}' + '\\}[^C]' # dodanie warunku na brak wystąpienia
    !grep '^{var}' {fl} | head
```

```python
# zadanie 6
var = 'UCUUUGGUAUCCUAG'
fl = 'pliki2/mature.fa'

# pierwsze wyszukanie
!grep '{var}' {fl} 

# wiersze przed trafieniem
!grep -B1  '{var}' {fl}

# ??? co tutaj powinno się zrobić 
```

```python
# zad 8 
var = 's/day/night/' # zmiana znaków 
fl = 'pliki2/geny'

!sed '{var}' {fl}
```

```python
# zad 9
new_file = '1, 6, 1749' #UWAGA spacji ni
sed_com = 's/([0-9]),[[:space:]]([0-9])/\\2,\\1/' # stosuje opcje -E 

# chcemy zamienić wiersze na kolumny 
!echo {new_file} | sed -E '{sed_com}'

# inne meta znaki
```

```python
# zad 11
fl = 'pliki2/geny'
gr_cop = "grep 'night'"

!sed 's/day/night/' {fl} | {gr_cop}
!sed 's/day/night/1' {fl} | {gr_cop}
!sed 's/day/night/2' {fl} | {gr_cop}
!sed 's/day/night/2g' {fl} | {gr_cop}
!sed 's/day/night/3' {fl} | {gr_cop}
```

```python
# zadanie 12
fl = 'pliki2/geny'

!sed -n 's/protein/peptide/p' {fl}
```
