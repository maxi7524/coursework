# Część 1: Wprowadzenie o biblioteki Polars

[**Polars**](https://docs.pola.rs/) to nowoczesna i ogólnodostępna biblioteka do analizy danych zaprojektowana z myślą o wysokiej wydajności i elastyczności. Bazuje na języku Rust oraz specyfikacji Apache Arrow, co pozwala na przetwarzanie danych w formacie kolumnowym z wykorzystaniem technik optymalizacji współczesnych procesorów. Polars jest szczególnie przydatny w sytuacjach, gdy mamy do czynienia z dużymi zbiorami danych lub wymagamy szybkich i wydajnych obliczeń.

### W porównaniu do bardziej tradycyjnych narzędzi, takich jak Pandas, Polars oferuje:

* Wyjątkową wydajność: Dzięki zastosowaniu języka Rust, Polars wykorzystuje możliwości nowoczesnych procesorów, w tym optymalizacje SIMD (Single Instruction, Multiple Data), co czyni go jedną z najszybszych bibliotek do analizy danych.
* Obsługę dużych zbiorów danych: Polars wspiera zarówno przetwarzanie w pamięci (in-memory), jak i tryb przetwarzania strumieniowego (streaming), co pozwala na pracę z danymi, które nie mieszczą się w pamięci RAM.
* Elastyczność: Dzięki podejściu „eager” i „lazy” (więcej na ten temat poniżej), użytkownik może dopasować sposób przetwarzania danych do konkretnego problemu.

### Eager i Lazy API
Polars udostępnia dwa różne style pracy z danymi:

* Eager API: Operacje są wykonywane natychmiast po ich wywołaniu. Ten tryb jest intuicyjny i przypomina pracę z Pandas. Idealny do eksploracji danych.
* Lazy API: Operacje są rejestrowane w tzw. grafie zapytań, a ich wykonanie jest odkładane do momentu, gdy wynik jest naprawdę potrzebny. Lazy API pozwala na automatyczną optymalizację zapytań, co jest szczególnie przydatne przy pracy z dużymi zbiorami danych.

### Kiedy warto używać Polars?
Polars to doskonały wybór w następujących przypadkach:

* Analizujesz bardzo duże zbiory danych, które wykraczają poza możliwości pamięci RAM.
* Potrzebujesz maksymalnej wydajności i chcesz uniknąć ograniczeń Pandas.
* Chcesz korzystać z zaawansowanych technik optymalizacji zapytań, aby przyspieszyć przetwarzanie.
* Pracujesz w środowiskach, gdzie czas odpowiedzi i efektywność są kluczowe, np. w aplikacjach czasu rzeczywistego.

### Zalety architektury Rust
Rdzeń Polars został napisany w języku Rust, co gwarantuje wysoką wydajność i bezpieczeństwo pamięci. Rust pozwala unikać typowych błędów związanych z zarządzaniem pamięcią, jednocześnie oferując funkcje, które maksymalizują wykorzystanie sprzętu.

## Struktura danych w bibliotece Polars

Polars oferuje trzy główne struktury danych:

* Series – reprezentuje jedną kolumnę danych o jednolitym typie.
* DataFrame – dwuwymiarowa tabela danych, składająca się z wierszy i kolumn.
* LazyFrame – struktura działająca w trybie leniwym, gdzie operacje są optymalizowane i wykonywane dopiero wtedy, gdy są naprawdę potrzebne.
Praca z LazyFrame jest szczególnie wydajna, gdy analizujemy duże zbiory danych, ponieważ biblioteka może automatycznie optymalizować zapytania.

## Filtracja danych
Filtracja danych to kluczowa operacja w analizie danych, pozwalająca na wybór tylko tych wierszy, które spełniają określone kryteria. W bibliotece Polars do filtrowania używamy metody `.filter()`, która działa na obiektach DataFrame. W połączeniu z funkcją `pl.col()` umożliwia precyzyjne odwołanie się do kolumny, na której chcemy przeprowadzić filtrowanie.



`.filter()` - to metoda, która stosuje podany warunek do wszystkich wierszy w DataFrame i zwraca nowy DataFrame zawierający tylko te wiersze, które spełniają kryterium.

```python
## Przykład - filtracja danych wg wieku
medical_df.filter( pl.col('Age') >= 30)
```

### Filtracja danych z wieloma parametrami

```python
## Proszę zwrócić uwagę na składnię, która jest zupełnie inna niż w Pandas

medical_df.filter( pl.col('Age') >= 30, pl.col('Smoker') == True)
```

## Dodawanie i usuwanie kolumn


### Dodawanie kolumn
Aby dodać nową kolumnę na podstawie istniejących, używamy metody `.with_columns()`. Metoda ta przyjmuje wyrażenia (`Expr`), które definiują nowe kolumny.

```python
## Przykład: dodanie nowej kolumny z wysokością w stopach

## 1 stopa = 1 cm / 30.48

medical_df = medical_df.with_columns(
    ( pl.col("Height (cm)")/30.48).round(2).alias("Height in feet")
    )
```

```python
## Zadanie: Proszę dodać nową kolumnę zawierającą wartość BMI

medical_df = medical_df.with_columns((pl.col("Weight (kg)")/((pl.col("Height (cm)")/100)**2)).round(2).alias("BMI"))
```

#### Dlaczego nie widzimy bezpośrednio wyniku grupowania?
* W bibliotece Polars do grupowania używa się metody `group_by()`, podobnie jak w Pandas. Jednak, podobnie jak w Pandas, samo `group_by()` nie wyświetla wyników – należy zastosować funkcję agregującą.
* Samo group_by() w Polars zwraca `obiekt grupujący (LazyGroupBy lub GroupBy)`, który musi zostać poddany agregacji, aby uzyskać wynik.

```python
## Grupowanie i obliczanie średniej wagi

medical_df.group_by('Gender').agg([
    pl.col("Weight (kg)").mean().alias("Average Weight (kg)")
])
```

```python
#### Multiple aggregation

medical_df.group_by("Gender").agg([
    pl.col("Weight (kg)").mean().alias("Average Weight (kg)"),
    pl.col("Cholesterol Level").mean().alias("Average Cholesterol Level")
])
```

```python
## Zadanie: Sprawdź czy osoby palące mają wyższy średni poziom cholesterolu?
medical_df.group_by("Smoker").agg([
    (pl.col("Cholesterol Level").sum()/pl.col("Cholesterol Level").count()).alias("Sredni cholesterol")
])
```

### Wczytanie danych z poszczególnych eksperymentów do większego obiektu Data Frame
Poniższy kod ładuje dane z plików CSV znajdujących się w folderze PyConPL24_Pandas_Polars/raw_data_robot. Proces ten wykorzystuje lazy loading z biblioteki Polars, co pozwala na efektywne zarządzanie dużymi zestawami danych. Każdy plik zostaje oznaczony dodatkową kolumną wskazującą jego ścieżkę, a wszystkie pliki są scalane w jeden duży DataFrame.

```python
# Najpierw tworzymy plan, a właściwie listę planów - listę działań krok po kroku, które zostaną wykonane tylko wtedy, kiedy sobie tego życzymy.
# Przygotowujemy wszystkie pliki ze wskazanego katalogu i "skanujemy" (scan_csv). Na tym etapie przygotowujemy pliki do odczytu, ale nic się nie dzieje. scan_csv inicjuje korzystanie z LazyFrame.
#

import glob

queries = []
for filepath in glob.glob("PyConPL24_Pandas_Polars/raw_data_robot/*.csv"): # Pobieramy wszystkie pliki z folderu
    df = pl.scan_csv(filepath, separator="\t", null_values="nan") # Używamy LazyFrame, aby stworzyć plan wczytywania danych
    df = df.with_columns(pl.lit(str(filepath)).alias("file_path")) # Dodajemy kolumnę z informacją o ścieżce pliku
    queries.append(df) # Dodajemy każdy plan (lazyframe) do listy
# let's load all these files into a large DataFrame

import glob

queries = []
for filepath in glob.glob("PyConPL24_Pandas_Polars/raw_data_robot/*.csv"): # we take all the files in the folder
    df = pl.scan_csv(filepath, separator="\t",null_values="nan") # we will use "LazyFrame" to create a plan for loading the data
    df = df.with_columns(pl.lit(str(filepath)).alias("file_path")) # for each file we add a column with the file path
    queries.append(df) # we add each such plan/ lazyframe to a list

plan = pl.concat(queries) # we concatenate all the plans into one
multiple_experiments = plan.collect() # we turn the plan into reality and load the data

plan = pl.concat(queries) # Łączymy wszystkie plany w jeden

## To tego momentu program jeszcze niczego nie wykonał - na razie zaplanował pracę. Dopiero w poniższym kroku coś się dzieje.
##  Polars bardzo efektywnie wykorzystuje moce obliczeniowe i może wykonywać wiele czynności równolegle.

multiple_experiments = plan.collect() # Realizujemy plan i wczytujemy dane

```

### Zapis wyników do pliku - porównanie formatów csv i parquet

Zapis wyników do pliku:
* `write_parquet:`
Zapisuje dane w formacie Parquet, który jest binarnym formatem zoptymalizowanym pod kątem efektywnego przechowywania i odczytu dużych zbiorów danych.
Jest bardziej wydajny niż format CSV pod względem rozmiaru pliku i szybkości odczytu/zapisu.

### Aby zrozumieć zachowanie balonu, musimy obliczyć jego objętość

Do tego celu potrzebujemy skorzystać z przepływu gazu (flow rate). Objętość oblicza się jako sumę przepływu gazu w czasie. Można to zapisać matematycznie w następujący sposób:

$$
Volume = \sum_{t=0}^{T_{end}} ({gas\_flow\_{si}} * \Delta t)
$$

Where:
- $Volume$  to całkowita objętość gazu,
- $T_{end}$ to ostatni krok czasowy,
- $gas\_flow_{si}$ to przepływ gazu w każdym kroku czasowym $t$,
- $\Delta t$ to odstęp czasowy między kolejnymi krokami.

# Zacznijmy od obliczenia tego dla jednego eksperymentu:
Musimy:

1. przefiltrować dane dla jednego eksperymentu,
2. obliczyć różnicę czasu
3. obliczyć przyrost objętości dla każdego kroku czasowego,
4. zsumować te wartości.

```python
single_experiment = multiple_experiments.filter(pl.col("file_path") == "PyConPL24_Pandas_Polars/raw_data_robot/Bt_2.csv")

px.scatter(single_experiment, x="time",y=["gas_flow"]) # Widzimy, że przepływ gazu nieco się waha i potrzebuje czasu, aby się rozpocząć.
```

Polars oferuje wiele wbudowanych funkcji, a najbardziej "pandasowy" sposób na rozwiązanie tego problemu będzie polegał na utworzeniu kilku nowych serii: różnicy czasu (time delta), następnie przyrostu objętości (volume delta), a na końcu ich zsumowaniu.

```python
time_deltas = single_experiment["time"].diff() # Obliczamy różnice czasu (time deltas) między kolejnymi pomiarami.

time_deltas.plot.hist() # Różnice w czasie są dość stabilne, ale nie idealne
```

```python
volume_deltas = single_experiment["gas_flow"] * time_deltas # Proszę zamienić kolumnę na 'gas_flow_si' !
px.line(volume_deltas.to_frame()) # Widzimy, że przepływ nieco się waha i potrzebuje czasu, aby się ustabilizować.
#volume_deltas.plot.plot() # Przepływ jest zaskakująco stabilny
```

```python
# Teraz używamy kumulatywnej sumy (cumulative sum), aby obliczyć sumę przyrostów objętości (volume deltas) do danego punktu.
volumes = volume_deltas.cum_sum()

# Dodajemy obliczoną objętość do oryginalnych danych.
single_experiment = single_experiment.with_columns(volumes.alias("volume"))

px.line(single_experiment, x="time",y=["volume"])
#single_experiment.plot(x = "time", y = ["volume"]) # Widzimy, że objętość zwiększa się w czasie w niemal liniowy sposób (po początkowej fazie)
```

```python
# Możemy to również zrobić w bardziej "polarsowy" sposób, korzystając z wyrażeń (expressions) - tego nie ma w Pandas.

# Tworzymy wyrażenie, które oblicza kumulatywną sumę (cumulative sum).
calculate_delta_volumes = (  pl.col("time").diff() * pl.col("gas_flow")).cum_sum().alias("volume")

single_experiment = single_experiment.with_columns(calculate_delta_volumes)
px.line(single_experiment, x="time",y=["volume"])
#single_experiment.plot(x = "time", y = ["volume"]) # Widzimy, że objętość zwiększa się w czasie w niemal liniowy sposób (po początkowej fazie).
```

Teraz możemy zrobić to samo dla wszystkich eksperymentów, bez użycia pętli, korzystając z tego samego wyrażenia i metody` over,` za pomocą pojedynczego polecenia!

```python
# Obliczamy objętości dla wszystkich eksperymentów.

multiple_experiments = multiple_experiments.with_columns(calculate_delta_volumes.over("file_path"))
```
