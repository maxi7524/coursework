### Implemented Functions

- `gradient_descent`  
  Function that computes the minimum of a given symbolic function using gradient descent.

- `plot_3d`  
  Visualizes the loss surface in 3D with arrows indicating the direction of updates.

- `plot_3d_on_2d`  
  Displays the 2D projection of the optimization path on the domain.


```python
# basics 
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import sympy as sp
# task 4
from sklearn.linear_model import LinearRegression
# task 5
from scipy.optimize import minimize
```

```python
def gradient_descent(x0, grad, learning_rate=0.01, max_iter=10000, tol=1e-6):
    # keeping data
    history = [x0]
    # creating recurrence 
    x = np.array(x0)
    for _ in range(max_iter):
        next_x = x - learning_rate * np.array(grad(*x.tolist()))
        # checking condition
        if np.linalg.norm(next_x - x) < tol:
            break
        history.append(next_x)
        x = next_x
    return np.array(history)
```

```python

def plot_3d(func: callable, func_g: list, interval: list, func_name: str):
    # local variables
    a, b = interval
    dens = int(np.ceil(b - a) * 10)
    # initiating 
    fig ,ax = plt.subplots(subplot_kw={'projection': '3d'})
    arg_range = np.linspace(a, b, dens)
    xx, yy = np.meshgrid(arg_range, arg_range)
    z = func(xx, yy)

    # plotting 
    ax.plot_surface(xx, yy, z, cmap='viridis', alpha=0.7)

    # gradient trajectory 
    ax.scatter(func_g[:, 0], func_g[:, 1], func(func_g[:, 0], func_g[:, 1]), color='red')

    # arrows indicating move 
    for i in range(len(func_g) - 1):
        # values 
        x, y, x_n, y_n = *func_g[i, :], *func_g[i + 1, :]
        ax.quiver(x, y, func(x, y),
                x_n - x, y_n - y, func(x_n, y_n) - func(x, y), 
                color='red', arrow_length_ratio=0.1, alpha=0.7)

    # description 
    ax.set_title(f"Gradient Descent on {func_name}")
    ax.set_xlabel('x')
    ax.set_ylabel('y')
    ax.set_zlabel('f(x, y)')
    plt.show()
```

```python
def plot_3d_on_2d(func: callable, func_g: list, interval: list, lvl: float, func_name: str):
    # creating local variables 
    a, b = interval
    dens = int(np.ceil(b-a) * 10)
    # initiating 
    arg_range = np.linspace(a, b, dens)
    xx, yy = np.meshgrid(arg_range, arg_range)
    z = func(xx, yy)

    # level set 
    z_max, z_min = z.max(), z.min()
    levels = int(np.ceil(z_max - z_min) * lvl)
    contour = plt.contour(xx, yy, z, levels=levels, cmap='viridis')
    plt.clabel(contour, inline=True, fontsize=8)

    # gradient trajectories 
    plt.scatter(func_g[:, 0], func_g[:, 1], color='red', label='Gradient Path')
    # gradient arrows 
    for i in range(len(func_g) - 1):
        x, y, x_n, y_n = *func_g[i, :], *func_g[i+1, :]
        plt.quiver(x, y, 
                x_n - x, y_n - y,
                angles='xy', scale_units='xy', scale=1.1, color='red', alpha=0.9)

    # display 
    # plt.colorbar(contour)
    plt.title(f"Gradient Descent on {func_name} (Contour Plot)")
    plt.xlabel('x')
    plt.ylabel('y')
    plt.grid(True)
    plt.show()
```

# Task 1

Write an algorithm for finding the minimum by the gradient descent method.
Present it for the function $f(x)=x^2$, starting from $x=2$ and illustrate it by marking the first 5 iterations on the figure.



```python
# parameters 
lr = 0.15
n_iter = 5

# calculating function and its derivative 
x = sp.Symbol('x')
func_1 = x**2
func_1_d = sp.diff(func_1)
func_1 = sp.lambdify(x, func_1)
func_1_d = sp.lambdify(x, func_1_d)
# plotting function 
x_domain = np.linspace(-3, 3, 100)
plt.plot(x_domain, func_1(x_domain), label='$f(x) = x^2$', color='blue')

traj1 = gradient_descent([2], func_1_d, learning_rate=lr, max_iter=n_iter)
plt.scatter(traj1[:, 0], traj1[:, 0]**2, color='red')
plt.quiver(traj1[:-1, 0], traj1[:-1, 0]**2, 
           traj1[1:, 0] - traj1[:-1, 0], 
           (traj1[1:, 0]**2) - (traj1[:-1, 0]**2),
           angles='xy', scale_units='xy', scale=1, color='red')
```

# Task 2

Write an algorithm for finding the minimum by the gradient descent method.
Present it for the function $f(x,y)=x^2-2x+y^2$, starting from $(x,y)=(2,2)$ and illustrate it by marking the first 5 iterations on the figure.

```python
# parameters 
lr = 0.1
n_iter = 5

# calculating functions 
x, y = sp.symbols('x, y')
func_2 = x**2 - 2*x + y**2
func_2_d = sp.diff(func_2, x), sp.diff(func_2, y)
func_2 = sp.lambdify((x, y), func_2)
func_2_d = sp.lambdify((x, y), func_2_d)
func_2_d_values = gradient_descent(
        [-2, 2], func_2_d, 
        learning_rate=lr, max_iter=n_iter
        )

plot_3d(func_2, func_2_d_values, [-2, 2], '$f(x, y) = x^2 - 2x + y^2$')
plot_3d_on_2d(func_2, func_2_d_values, [-2, 2], 1, '$f(x, y) = x^2 - 2x + y^2$')
```

# Task 3


Write an algorithm for finding the minimum by the gradient descent method.
Present it for the function $f(x,y)=x^2-y^2$, starting from $(x,y)=(5,1)$ and illustrate it by marking the first 5 iterations on the figure.

Create:
  * 2D conture plot   (e.g. https://jakevdp.github.io/PythonDataScienceHandbook/04.04-density-and-contour-plots.html)
  * 3D 
  
  What happens when we change a starting point into $(5,0)$?

### Solution

**`Goal`**  
We want to see that the behavior of gradient descent depends on the starting point.

**`Methodology`**  
We use the previously defined `gradient_descent` function and apply it to two starting points: $(5, 1)$ and $(5, 0)$.  
Note: In the second case, we use $(5, -1)$ instead of $(5, 0)$ for better plot visibility and alignment with the plot's orientation.

**`Results & Discussion`**  
- In the first plot, we observe that the method diverges to $\infty$, as the $y$-axis does not lead to a minimum.
- In the second plot, the algorithm converges toward $(0, 0)$ since it starts along the $x$-axis, which aligns with the function's minimizing direction.



```python
# parameters
lr = 0.1
x0 = [5, 1]
n_iter = 5

# creating functions 
func_3 = x**2 - y**2 
func_3_d = sp.diff(func_3, x), sp.diff(func_3, y)
func_3 = sp.lambdify((x, y), func_3)
func_3_d = sp.lambdify((x, y), func_3_d)
for x0 in [[5,-1], [5,0]]:
    func_3_d_values = gradient_descent(x0, func_3_d, lr, n_iter)
    plot_3d_on_2d(func_3, func_3_d_values, [-7, 7], 0.4, '$f(x, y) = x^2 - y^2$')
    plot_3d(func_3, func_3_d_values, [-7, 7], '$f(x, y) = x^2 - y^2$')
```

# Task 4
Define a square error function:

$$
\mathrm{compute\_error}(a,b) =  \sum_{i=1}^{n} ( y_i - (ax_i+b) )^2.
$$

and use a gradient descent method from Task 2 to find a minimum.


### Solution part
**`goal`**
We want to find minimum for RSS function so we can see how algorithm behaves (here we have deterministic solution so we can compare results) 

**`Method`**: 
- i create sympy function, to be able to differentiate function 
- i calculate coefficients using my implemented function

**`Results & Discussion`**
- we see that our algorithm converges properly to proper value difference is small



```python
# parameters 
x0 = [2, 2] # starting point 


# generated values 
f = lambda x: (x**2)
x = np.array([.2, .5, .8, .9, 1.3, 1.7, 2.1, 2.7])
y = f(x) + np.random.randn(len(x))

# creating RSS functions 
a, b = sp.symbols('a, b')
RSS = sp.simplify(sum((y - ((a*x) + b))**2))
RSS_d = [sp.diff(RSS, a), sp.diff(RSS, b)]
RSS = sp.lambdify((a, b), RSS)
RSS_d = sp.lambdify((a, b), RSS_d)

# calculating minimum 
gd_val = gradient_descent(x0, RSS_d)
gd_min = gd_val[-1]

# calculating minimum by using linear model
lm = LinearRegression()
lm.fit(x.reshape(-1, 1), y)
lm_min = lm.coef_.item(), lm.intercept_


print(f'GD min = {gd_min} with {len(gd_val)} iterations\nLM_min = {lm_min}')
```

* $\mathrm{compute\_error}(a,b)$ (by parameters $a,b$)

Minimalized the function $\mathrm{compute\_error}(a,b)$ by <tt>optimize.fmin_cg</tt> (by parameters $a,b$)

### Solution 

**`Goal`**  
Apply a built-in optimization algorithm from `scipy.optimize` to minimize a symbolic error function and compare its performance to our custom implementation.

**`Methodology`**  
- Created lambda functions compatible with `optimize.fmin_cg` (i.e., first parameter must be an \(n\)-dimensional array).
- Used `scipy.optimize.minimize` with method `'CG'` (Conjugate Gradient), which offers a cleaner and more general interface.

**`Results & Discussion`**  
- Both methods yield the same final coefficients.
- The `scipy` implementation converges much faster (7 iterations vs. 357), demonstrating the advantage of using highly optimized libraries.


```python
# parameters 
x0 = np.array([2, 2])
# symbols
f = lambda x: (x[0] + 1)**2 + x[1]**2
com_err = lambda b: sum((y - (x * b[0] + b[1]))**2)

# optimizing 
opt_obj_1 = minimize(f, x0, method='CG')
opt_obj_2 = minimize(com_err, x0, method='CG')

print(opt_obj_1.x, '\n--------------------------------\nRSS comparison \n', opt_obj_2.x, gd_min, '\nIterations\n', opt_obj_2.nit, 'vs', len(gd_val))
```

# Task 6
Define an error function::

$$
\mathrm{compute\_error\_1}(a,b) =  \sum_{i=1}^{n} |y_i - (ax_i+b)|.
$$

and find minimum. Compare it with the minimum of error square function ($\mathrm{compute\_error}(a,b)$).


### Solution

**`Goal`**  
Minimize a cost function based on absolute error and compare it to a solution using squared error (RSS).

**`Methodology`**  
- Used `SymPy` to symbolically define and minimize the absolute error function.
- Used `LinearRegression` from `sklearn` to minimize RSS for comparison.

**`Results & Discussion`**  
- The absolute error solution is more resistant to outliers, as the influence of each data point increases linearly with the error. This makes the model fit closer to the majority of the data.  
- In contrast, the RSS approach squares the residuals, which amplifies the effect of extreme values. As a result, the model may shift toward outliers in order to minimize large errors.

```python
# parameters 
x0 =[2, 2]
com_err_2 = lambda b: np.sum(abs(y - (b[0] * x + b[1])))
f = lambda x: (x**2)

# data
x = np.array([.2, .5, .8, .9, 1.3, 1.7, 2.1, 2.7])
y = f(x) + np.random.randn(len(x))

# plotting values 
x_tr = np.linspace(0., 3, 200)
y_tr = f(x_tr)

# linear model 
lr = LinearRegression()
lr.fit(x[:, np.newaxis], y);
print(lr.coef_)
print(lr.intercept_)

f_lr = lambda x: lr.coef_ * x +lr.intercept_

x_f_lr = np.linspace(0., 3, 200)
y_f_lr = f_lr(x_tr)
plt.figure(figsize=(6,6));
axes = plt.gca()
axes.set_xlim([0,3])
axes.set_ylim([0,8])
#plt.plot(x_tr, y_tr, '--k');
plt.plot(x_f_lr, y_f_lr, 'g', label='RSS');

# plotting abs approximation 
# minimization 
opt_com_err_2 = minimize(com_err_2, x0, method='CG')
opt_com_err_2.x
a_, b_ = opt_com_err_2.x
f_lr_abs = a_ * x_f_lr + b_
plt.plot(x_f_lr, f_lr_abs, color='blue', label='absolute')

plt.plot(x, y, 'ok', ms=10);
plt.legend()
plt.show()
```
