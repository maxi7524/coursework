Document description 

General gradient descent, used in every exercise
- it takes all parameters
- it iterates until certain point difference is by passed / certain number of iterations 
- it handles both, 1 dimensional and multidimensional case

```python
def gradient_descent(x0, grad, learning_rate=0.01, max_iter=100, tol=1e-6):
    # keeping data
    history = [x0]
    # creating recurrence 
    x = np.array(x0)
    for _ in range(max_iter):
        next_x = x - learning_rate * np.array(grad(*x.tolist()))
        # checking condition
        if np.linalg.norm(next_x - x) < tol:
            break
        history.append(next_x)
        x = next_x
    return np.array(history)
```

```python

def plot_3d(func: callable, func_g: list, interval: list, func_name: str):
    # local variables
    a, b = interval
    dens = int(np.ceil(b - a) * 10)
    # initiating 
    fig ,ax = plt.subplots(subplot_kw={'projection': '3d'})
    arg_range = np.linspace(a, b, dens)
    xx, yy = np.meshgrid(arg_range, arg_range)
    z = func(xx, yy)

    # plotting 
    ax.plot_surface(xx, yy, z, cmap='viridis', alpha=0.7)

    # gradient trajectory 
    ax.scatter(func_g[:, 0], func_g[:, 1], func(func_g[:, 0], func_g[:, 1]), color='red')

    # arrows indicating move 
    for i in range(len(func_g) - 1):
        # values 
        x, y, x_n, y_n = *func_g[i, :], *func_g[i + 1, :]
        ax.quiver(x, y, func(x, y),
                x_n - x, y_n - y, func(x_n, y_n) - func(x, y), 
                color='red', arrow_length_ratio=0.1, alpha=0.7)

    # description 
    ax.set_title(f"Gradient Descent on {func_name}")
    ax.set_xlabel('x')
    ax.set_ylabel('y')
    ax.set_zlabel('f(x, y)')
    plt.show()
```

```python
def plot_3d_on_2d(func: callable, func_g: list, interval: list, lvl: float, func_name: str):
    # creating local variables 
    a, b = interval
    dens = int(np.ceil(b-a) * 10)
    # initiating 
    arg_range = np.linspace(a, b, dens)
    xx, yy = np.meshgrid(arg_range, arg_range)
    z = func(xx, yy)

    # level set 
    z_max, z_min = z.max(), z.min()
    levels = int(np.ceil(z_max - z_min) * lvl)
    contour = plt.contour(xx, yy, z, levels=levels, cmap='viridis')
    plt.clabel(contour, inline=True, fontsize=8)

    # gradient trajectories 
    plt.scatter(func_g[:, 0], func_g[:, 1], color='red', label='Gradient Path')
    # gradient arrows 
    for i in range(len(func_g) - 1):
        x, y, x_n, y_n = *func_g[i, :], *func_g[i+1, :]
        plt.quiver(x, y, 
                x_n - x, y_n - y,
                angles='xy', scale_units='xy', scale=1.1, color='red', alpha=0.9)

    # display 
    # plt.colorbar(contour)
    plt.title(f"Gradient Descent on {func_name} (Contour Plot)")
    plt.xlabel('x')
    plt.ylabel('y')
    plt.grid(True)
    plt.show()
```

# Task 1

Write an algorithm for finding the minimum by the gradient descent method.
Present it for the function $f(x)=x^2$, starting from $x=2$ and illustrate it by marking the first 5 iterations on the figure.



```python
# parameters 
lr = 0.15
n_iter = 5

# calculating function and its derivative 
x = sp.Symbol('x')
func_1 = x**2
func_1_d = sp.diff(func_1)
func_1 = sp.lambdify(x, func_1)
func_1_d = sp.lambdify(x, func_1_d)
# plotting function 
x_domain = np.linspace(-3, 3, 100)
plt.plot(x_domain, func_1(x_domain), label='$f(x) = x^2$', color='blue')

traj1 = gradient_descent([2], func_1_d, learning_rate=lr, max_iter=n_iter)
plt.scatter(traj1[:, 0], traj1[:, 0]**2, color='red')
plt.quiver(traj1[:-1, 0], traj1[:-1, 0]**2, 
           traj1[1:, 0] - traj1[:-1, 0], 
           (traj1[1:, 0]**2) - (traj1[:-1, 0]**2),
           angles='xy', scale_units='xy', scale=1, color='red')
```

# Task 2

Write an algorithm for finding the minimum by the gradient descent method.
Present it for the function $f(x,y)=x^2-2x+y^2$, starting from $(x,y)=(2,2)$ and illustrate it by marking the first 5 iterations on the figure.

```python
# parameters 
lr = 0.1
n_iter = 5

# calculating functions 
x, y = sp.symbols('x, y')
func_2 = x**2 - 2*x + y**2
func_2_d = sp.diff(func_2, x), sp.diff(func_2, y)
func_2 = sp.lambdify((x, y), func_2)
func_2_d = sp.lambdify((x, y), func_2_d)
func_2_d_values = gradient_descent(
        [-2, 2], func_2_d, 
        learning_rate=lr, max_iter=n_iter
        )

plot_3d(func_2, func_2_d_values, [-2, 2], '$f(x, y) = x^2 - 2x + y^2$')
plot_3d_on_2d(func_2, func_2_d_values, [-2, 2], 1, '$f(x, y) = x^2 - 2x + y^2$')
```

Define an error function::

$$
\mathrm{compute\_error\_1}(a,b) =  \sum_{i=1}^{n} |y_i - (ax_i+b)|.
$$

and find minimum. Compare it with the minimum of error square function ($\mathrm{compute\_error}(a,b)$).


```python
import numpy as np
import matplotlib.pyplot as plt
import sklearn.linear_model as lm

f = lambda x: (x**2)


x = np.array([.2, .5, .8, .9, 1.3, 1.7, 2.1, 2.7])
y = f(x) + np.random.randn(len(x))

x_tr = np.linspace(0., 3, 200)
y_tr = f(x_tr)

# model.
lr = lm.LinearRegression()
lr.fit(x[:, np.newaxis], y);
print(lr.coef_)
print(lr.intercept_)

f_lr = lambda x: lr.coef_ * x +lr.intercept_

x_f_lr = np.linspace(0., 3, 200)
y_f_lr = f_lr(x_tr)
plt.figure(figsize=(6,6));
axes = plt.gca()
axes.set_xlim([0,3])
axes.set_ylim([0,8])
#plt.plot(x_tr, y_tr, '--k');
plt.plot(x_f_lr, y_f_lr, 'g');
plt.plot(x, y, 'ok', ms=10);
plt.show()
```
