x[a, b] zwraca nam wartość x'owej współrzędnej na polu a, b
y[a, b] zwraca nam wartośc y'owej współrzędnej na polu a, b

x = x[0,0], x[1,0] ...\ x[0, n], x[1, n]
x = 0, 1 ... 0, 1

x 

w przypadku większej ilości zmiennyh dostaniemy macierze których już nie będziemy w stanie zobrazować. 

```python
a, b, freq = 0, 3, 4
x = np.linspace(a, b, freq)
y = np.linspace(a, b, freq)

xz, yz = np.meshgrid(x, y)
f = xz**3 + yz*10
xz, yz, f
```

```python
plt.figure(figsize=(4, 1.5))
plt.pcolormesh(xz, yz, f)
plt.colorbar()
plt.show()

```
