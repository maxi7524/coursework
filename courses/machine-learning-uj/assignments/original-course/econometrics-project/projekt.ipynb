```python
# dropping linear values``
X_t = X.drop(['Diameter'], axis=1)

# decoding categorical data 'Sex'
categorical_data = X_t.select_dtypes(include=['object', 'category']).columns
X_encoded = pd.get_dummies(X_t, drop_first=True)
cols_encd = list(X_encoded.columns)
X_encoded = X_encoded[cols_encd[-2:] + cols_encd[:-2]]


# splitting data
# X.drop(['Sex'], axis=1)
X_train, X_test, y_train, y_test = train_test_split(X_encoded, y, test_size=0.2, random_state=seed)

# calculatting model 
model = make_pipeline(
    StandardScaler(),
    # PCA(6),
    LinearRegression()
    )
model.fit(X_train, y_train)
model.score(X_test, y_test), 'Shucked_weight'

# X_encoded

# ważne tej shave jest bardzo ważny 
```
