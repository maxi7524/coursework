```python
# basics 
import pandas as pd
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt
# models 
from sklearn.linear_model import LogisticRegression, LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.neighbors import KNeighborsClassifier
# basics 
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.metrics import classification_report, confusion_matrix
```

Generating data

```python
N=1000

df1 = 2*pd.DataFrame(np.random.randn(N, 3), columns=['X', 'Y','C'])
df2 = 20+10*pd.DataFrame(np.random.randn(N, 3), columns=['X', 'Y','C'])
df1['C'] = 1
df2['C'] = 0
df=pd.concat((df1, df2), ignore_index=True)
```

```python
X_train, X_test, y_train, y_test = train_test_split(df[['X', 'Y']], df['C'], test_size=0.2)

# knn model
model_KNN = KNeighborsClassifier()
model_KNN.fit(X_train, y_train)
predict_KNN = model_KNN.predict(X_test)

# linear regression
model_lin = LinearRegression()
model_lin.fit(X_train, y_train)
predict_lin = (model_lin.predict(X_test) >= 0.5).astype(int)

# logistic regression
model_regr = LogisticRegression()
model_regr.fit(X_train, y_train)
predict_regr = model_regr.predict(X_test)
```

Graphs 

```python
# 4. Plot decision boundaries
def plot_decision_boundary(model, model_name, thresholding=False):
    # Create grid
    h = 0.2
    x_min, x_max = df['X'].min() - 1, df['X'].max() + 1
    y_min, y_max = df['Y'].min() - 1, df['Y'].max() + 1
    xx, yy = np.meshgrid(np.arange(x_min, x_max, h),
                         np.arange(y_min, y_max, h))
    grid_points = np.c_[xx.ravel(), yy.ravel()]
    
    # Predict on grid
    if thresholding:  # for linear regression
        Z = (model.predict(grid_points) >= 0.5).astype(int)
    elif hasattr(model, "predict"):
        Z = model.predict(grid_points)
    else:
        raise ValueError("Model does not have predict method.")
    
    Z = Z.reshape(xx.shape)

    # Plot
    plt.figure(figsize=(7,6))
    plt.contourf(xx, yy, Z, alpha=0.3, cmap='bwr')
    plt.scatter(X_test['X'], X_test['Y'], c=y_test, cmap='bwr', edgecolor='k', s=20)
    plt.title(f"Decision Boundary - {model_name}")
    plt.xlabel('X')
    plt.ylabel('Y')
    plt.grid(True)
    plt.show()

# 5. Visualize each
plot_decision_boundary(model_KNN, "K-Nearest Neighbors")
plot_decision_boundary(model_lin, "Linear Regression (Thresholded)", thresholding=True)
plot_decision_boundary(model_regr, "Logistic Regression")
```

Scores 

```python
print(classification_report(y_test, predict_KNN))
print(confusion_matrix(y_test, predict_KNN))
```

```python
print(classification_report(y_test, predict_lin))
print(confusion_matrix(y_test, predict_lin))
```

```python
print(classification_report(y_test, predict_regr))
print(confusion_matrix(y_test, predict_regr))
```
