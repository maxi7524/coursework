# Comparison of kernels
## Characterization of kernels
- Kernels give ous characterization about how we interpret realization of random variable
- their impact falls with rising probes (how to check it do to probe size) BUT they have impact on **tails**
## Methodology
- We set interval $[-1, 1]$ nad estimate how those function behave



```python
x = np.linspace(-1.5, 1.5, 400)

# Kernels definitions
def rectangular(x):
    return 0.5 * ((x >= -1) & (x <= 1))

def triangular(x):
    return (1 - np.abs(x)) * ((x >= -1) & (x <= 1))

def epanechnikov(x):
    return 0.75 * (1 - x**2) * ((x >= -1) & (x <= 1))

def biweight(x):
    return (15/16) * (1 - x**2)**2 * ((x >= -1) & (x <= 1))

def triweight(x):
    return (35/32) * (1 - x**2)**3 * ((x >= -1) & (x <= 1))

def cosine(x):
    return (np.pi/4) * np.cos(np.pi * x / 2) * ((x >= -1) & (x <= 1))

# Lista jąder
kernels = {
    "Rectangular": rectangular,
    "Triangular": triangular,
    "Epanechnikov": epanechnikov,
    "Biweight": biweight,
    "Triweight": triweight,
    "Cosine": cosine
}

# Rysowanie
plt.figure(figsize=(8,6))
for name, func in kernels.items():
    plt.plot(x, func(x), label=name)

plt.title("Comparison of kernels KDE")
plt.xlabel("x")
plt.ylabel("K(x)")
plt.legend()
plt.grid(True)
plt.show()
```
