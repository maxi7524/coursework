```r
# Loading data SwissLabor 
df = data(SwissLabor)
# finding GLM model 
# Dane
data("SwissLabor")

# probabilistic model
swiss_probit <- glm(participation ~ . + I(age^2),
                    data = SwissLabor,
                    family = binomial(link = "probit"))

# summary 
summary(swiss_probit)

# plot
plot(participation ~ age, data = SwissLabor, ylevels = 2:1,
     main = "Participation in market vs age",
     xlab = "Age", ylab = "Participation $[0, 1]$")
```

margin effect

to jest jeszcze do przerobienia, ponieważ mozna na różny sposób uśredniać te dane, pomyśleć nad taką oceną jakościową modelu !!!

```r
## Discrete 

# mean density in prediction points
fav <- mean(dnorm(predict(swiss_probit, type = "link")))

# marginal effect
marginal_effects <- fav * coef(swiss_probit)
print("average effect on jth regressor ")
print(fav)
print(marginal_effects) # this command print special characters

## Continuos

# mean value for continuous random variables
av <- colMeans(SwissLabor[ , -c(1,7)])  # bez participation i foreign

# We create two sets
av <- data.frame(rbind(swiss = av, foreign = av),
                 foreign = factor(c("no","yes")))

# Predictions
av_link <- predict(swiss_probit, newdata = av, type = "link")
av_dens <- dnorm(av_link)

# Margin effect for foreign
print("foreign")
print(av_dens["swiss"] * coef(swiss_probit)[-7])

# does it mean does per change on this value how probability is increasing ? 


# important this technique allow ous to check if DATA are stable !!! we can have stable coeffs but if data are unstable, model predictions will look more random 
```

```r
# Model with constant
swiss_probit0 <- update(swiss_probit, formula = . ~ 1)

# Pseudo-R²
pseudo_R2 = 1 - as.vector(logLik(swiss_probit) / logLik(swiss_probit0))
print(pseudo_R2)
```

```r
# Deviance
deviance(swiss_probit)

# Pearson 
sum(residuals(swiss_probit, type = "pearson")^2) # this one normalizes 
sum(residuals(swiss_probit, type = "deviance")^2) # this one is without normalization
```
