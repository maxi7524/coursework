```r
options(encoding = "UTF-8")
Sys.setlocale("LC_ALL", "pl_PL.UTF-8")

# packages loader
use_package <- function(pkg) {
  if (!require(pkg, character.only = TRUE)) {
    install.packages(pkg)
    library(pkg, character.only = TRUE)
  }
}

# use_package('GGAlly')
use_package('ggplot2')
use_package('MASS')
```

```r
# loading data
data = read.table('skulls.dat.txt', header = TRUE)
# creating categorize data
data$Year = as.factor(data$Year)
head(data)
```

Dyskryminacja kwadratowa

```r
# creating model
qda_fit <- qda(Year ~ ., data = data)
# predicting data
qda_pred <- predict(qda_fit, data)
# creating error table
table(data$Year, qda_pred$class)
```

dyskryminacja LOO (leave one out)

```r
qda_fit_cv <- qda(Year ~ ., data = data, CV = TRUE)
table(data$Year, qda_fit_cv$class)
```

Funkcje liniowej dyskryminacji 

```r
# linear discrimination with 
lda_fit = lda(Year ~., data=data)
## information about priors, groups means, coefficients of LD 
lda_fit
## prediction
lda_pred = predict(lda_fit, data)
## tabela  
table(data$Year, lda_pred$class)

# linear discrimination with validation 
lda_fit_cv = lda(Year ~ ., data=data, CV=TRUE)
table(data$Year, lda_fit_cv$class)
```

```r
# plots of discriminated functions
data$LD1 = lda_pred$x[,1]
data$LD2 = lda_pred$x[,2]

# plot of two first functions
ggplot(data, aes(x=LD1, y=LD2, color=Year)) + geom_point()
```

Fischer classifications 

```r
# fischer discriminant - same as LDA for two classes
fischer_fit = lda(Year~., data=data)
# coefs
fischer_fit$scaling
# functions values
fischer_scores = predict(fischer_fit)$x
# histograms 
ggplot(data.frame(score=fischer_scores[,1], group=data$group), aes(x=score, fill=Year)) + geom_density(alpha=0.5)
```
