# Rozmiary i różnicę między klasami

```python
# przypadek, dla próbny $n$ elementowej
n= 5
dist = torch.distributions.Normal(0.0, 1.0)
sample_shape = torch.Size([n]) 
print(
dist.sample(sample_shape)[:4], "\n", dist.sample([n])[:4] # że pokazać że to jest to samo
)
# sprawdzamy rozmiary tego obiektu
sample_shape, dist.batch_shape, dist.event_shape
```

```python
# rozmiar batch'a  - ilość różnych rozkład - robimy wektor z rozkładów 
batch_size = 5
dist = torch.distributions.Normal(torch.zeros(2), torch.ones(2)) # UWAGA tutaj mamy DWIE RÓŻNE DYSTRYBUCJE 
dist.sample()
# sprawdzamy rozmiary, tutaj batch zero więc - sample to jest po prostu pojedyncza realizacja zmiennej losowej, 
(dist.batch_shape, dist.event_shape)
```

```python
# rozmiar zdarzenia (pojedyncza element próby ile ma zawierać zdarzeń) 
dist = torch.distributions.MultivariateNormal(torch.zeros(2), torch.eye(2)) # UWAGA TO JEST JEDNA DYSTRYBUCJA, WIELOWYMIAROWA - można stworzyć zależności między nimi
(dist.batch_shape, dist.event_shape)
# to nam mówi, że batch size - ilość różnych modeli to jeden (wymiar 0), ale event size - czyli rozmiar przestrzeni zmiennej losowej to dwa
```

Losowanie prób, oraz wyliczanie log proba (zobaczenie jak wektoryzacja wygląda)

```python
def simulate_coin_tosses(n, theta):
  '''
  I assume that n is int and theta is float
  '''
  # creating distribution 
  dist = torch.distributions.Bernoulli(torch.tensor(theta))
  # taking sample
  sample = dist.sample([n])
  return sample


sample = simulate_coin_tosses(50, 0.5)
dist = torch.distributions.Bernoulli(0.5)
dist.log_prob(sample)
```
