## 1. Matplotlib - podstawy wizualizacji z wykorzystaniem interfejsu pyplot

Matplotlib to kompleksowa biblioteka do tworzenia statycznych, interaktywnych i animowanych wizualizacji w Pythonie.  Jest niezwykle wszechstronna i może być używana zarówno do prostych wykresów, jak i skomplikowanych figur z wieloma podwykresami. W tej części omówimy podstawy wizualizacji z wykorzystaniem interfejsu matplotlib.pyplot. Zapewnia on domniemany (*implicit*) sposób tworzenia wykresów, podobny do tego znanego z MATLAB-a, gdzie operujemy na ostatnio utworzonym rysunku czy panelu. Każda funkcja pyplot zmienia rysunek, np. tworzy nowy rysunek, tworzy obszar rysowania, rysuje na nim linie i punkty, dodaje etykiety, itd. Jest on głównie przeznaczony do prostych przypadków generowania wykresów programistycznie.

W tym rozdziale przedstawimy podstawowe rodzaje wykresów dostępnych poprzez interfejs matplotlib.pyplot, ale osoby zainteresowane innymi rodzajmi wykresów oraz sposobami dodawania różnych elementów do generowanych rysunków odsyłamy do [dokumentacji](https://matplotlib.org/stable/api/pyplot_summary.html#module-matplotlib.pyplot).

Generalnie, funkcje graficzne matplotlib oczekują danych w postaci tablicy numpy.array lub obiektów, które można na nią przekształcić. Ramki danych pandas, czy nawet macierze numpy.matrix, mogą nie zawsze zadziałać w taki sposób jakiego oczekujemy, więc warto wcześniej przekształcić je na tablice numpy.

Rozpocznijmy od wyświetlenia prostego **wykresu linowego**:

```python
# Zaimportuj bibliotekę, a dokładje jej główny moduł pyplot
import matplotlib.pyplot as plt

# Przygotuj dane wejściowe
x = [1, 2, 3, 4]
y = [10, 20, 25, 20]

# Narysuj wykres liniowy
plt.plot(x, y) 
# Wyświetl wykres
plt.show() 
# Jupyter domyślnie woła show() na końcu każdej komórki, ale warto wywołać show() wprost
```

Tak jak oczekiwaliśmy został wyświetly wykres linowy przedstawiający nasze dane, ale brak na nim jakichkolwiek opisów osi czy tytułu. Uzupełnijmy zatem wykres o te elementy, a dodatkowo zmieńmy kolor, styl i grubość lini, a także dodajmy obserwacje.

```python
plt.plot(x, y, color='red', linestyle='dashed', linewidth=2, marker='o')
plt.xlabel('Oś X')
plt.ylabel('Oś Y')
plt.title('Przykładowy wykres liniowy')
plt.show()
```

Możemy również tworzyć wykresy liniowe z błędami pomiarów (error bars); przydatne przy pomiarach z replikatami, przedziałach ufności

```python
yerr = [2, 3, 2, 4]
plt.errorbar(x, y, yerr=yerr, fmt='o-', capsize=5)
plt.xlabel('Oś X')
plt.ylabel('Oś Y')
plt.title('Przykładowy wykres liniowy')
plt.show()
```

W dalszej części przedstawimy inne rodzaje wykresów. Podzielimy jednak każdy rysunek (Figure) na panele lub osie (Axes), gdzie pokażemy podstawowe sposoby zmiany właściwości wykresu. 

Inne popularne rodzaje wykresów to:
- **wykres punktowy** (scatter plot)

```python
# Stwórz nowy wykres o określonym rozmiarze w calach
plt.figure(figsize=(8,3))

# Dodaj lewą oś (panel) przy podziale rysunku 1x2
plt.subplot(1, 2, 1) #nrows, ncols, index
# Narysuj wykres punktowy
plt.scatter(x, y)


# Dodaj prawą oś (panel) przy tym samym podziale rysunku 1x2
plt.subplot(1, 2, 2)
# Narysuj ten sam wykres w kolorze zielonym i punktami kwadratowymi  
plt.scatter(x, y, c='green', marker='s')

# Wyświetl wykres
plt.show()
```

- **wykres kołowy** (pie plot)

```python
plt.figure(figsize=(8,3))

# Panel lewy
plt.subplot(1, 2, 1)
# Narysuj wykres kołowy
plt.pie(y)

# Panel prawy
plt.subplot(1, 2, 2)
# Narusyje ten sam wykres z etykietami i udziałem oraz zmianą kierunku i lokalizacji elementów wykresu
labels = ['A','B','C','D']
plt.pie(y, labels=labels, autopct='%.1f%%', startangle=90, counterclock=False)

plt.show()
```

- **wykres słupkowy** (bar plot)

```python
plt.figure(figsize=(8,3))

# Panel lewy
plt.subplot(1, 2, 1)
# Narysuj wykres słupkowy
plt.bar(x, y)

# Panel prawy
plt.subplot(1, 2, 2)
# Narysuj ten sam wykres dodając etykiety, zmieniając kolory i dodając grube krawędzie
plt.bar(x, y, tick_label=labels, color=['red','green'], edgecolor='black', alpha=0.5, linewidth=2)

plt.show()
```

- **histogram** (wykorzystywany do wizualizacji częstotliwości występowania wartości)

Tu dodatkowo przykład uzycia notacji Latex umieszczania do wyrażeń matematycznych w napisach.

```python
import numpy as np
np.random.seed(1024)

plt.figure(figsize=(5,4))

# Wygeneruj losowe dane z rozkładu normalnego
temp = np.random.normal(12, 10, 365)

plt.hist(temp, bins=20, color='gray', edgecolor='white')
plt.xlabel(r'Temperatura [$^\circ C$]')
plt.ylabel('Częstotliwość')
plt.title('Rozkład temperatury powietrza w roku')
plt.show()
```

- **boxplot** (wykres pudełkowy) - pokazuje rozkład, medianę, kwartyle, outliery; podobny cel ma **violinplot**

Tu dodatkowo użyto skali logarytmicznej na osi Y. Przy okazji przypomnienie co pokazuje boxplot:

- Prostokąt: Q1 (25%) do Q3 (75%) to jest środkowe 50% danych
- Linia w środku: mediana (Q2, 50%)
- Wąsy: zasięg danych (zazwyczaj 1.5 * IQR)
- Kropki/kółka: wartości odstające


```python
# Porównanie ekspresji genów w różnych warunkach
np.random.seed(1024)
data = [np.random.normal(100, 10, 200),
        np.random.normal(80, 20, 200),
        np.random.normal(90, 5, 200)]
plt.boxplot(data, labels=['control', 'knockout A', 'knockout B'])
#ustawienie skali logarytmicznej
plt.yscale("log") 
plt.ylabel('expression (log scale)')
plt.title('Gene expression comparison')
plt.show()
```

- **heatmap** - pozwala wizualizować macierze, np. porównanie korelacji ekspresji genów

```python
data = np.random.rand(10, 10)
plt.imshow(data, cmap='viridis', aspect='auto')
plt.colorbar(label='korelacja')
plt.show()
```

Znając podstawowe rodzaje wykresów narysujmy teraz bardziej złożone wykresy.

**Przykład 1.1**: Grupowany wykres słupkowy

```python
# Poziom ekspresji genów w różnych tkankach
tkanki = ['Mózg', 'Wątroba', 'Mięsień']
kontrola = [23, 35, 47]  # Poziom ekspresji w grupie kontrolnej
lek = [17, 28, 40]  # Poziom ekspresji po dodaniu leku

# Zdefiniuj szerokość słupka
szerokoscSłupka = 0.3

# Ustaw pozycję słupka na osi X
r1 = np.arange(len(kontrola))
r2 = [x + szerokoscSłupka for x in r1]

# Narysuj słupki dla obu grup
plt.bar(r1, kontrola, color='blue', width=szerokoscSłupka, 
        edgecolor='grey', label='Kontrola', alpha=0.8)
plt.bar(r2, lek, color='red', width=szerokoscSłupka, 
        edgecolor='grey', label='Lek', alpha=0.8)

# Dodaj etykiety, tytuł i legendę
plt.xlabel('Typ tkanki', fontweight='bold', fontsize=11)
plt.xticks([r + 0.5*szerokoscSłupka for r in range(len(kontrola))], tkanki, fontsize=10)
plt.ylabel('Poziom ekspresji genu (FPKM)', fontsize=11)
plt.title('Ekspresja genu BRCA1 w różnych tkankach', fontweight='bold', fontsize=12)
plt.legend(frameon=False, fontsize=10)
plt.grid(axis='y', alpha=0.3, linestyle='--')

plt.tight_layout()
plt.show()
```

**Przykład 1.2**: Wykres punktowy wraz regresją liniową oraz przekazywanie argumentów przez nazwę.

Przydatna jest możliwość **zapisywania wykresów** do różnych formatów

```python
# Przykładowe dane
np.random.seed(1024)
dane = {'wiek': np.random.randint(0,18,100),
        'col': np.random.choice(['r','b'], size=100)}
dane['wzrost'] = np.random.normal(10,8,100) + dane['wiek'] * 6 + 50
# Narysuj wykres punktowy
plt.scatter('wiek', 'wzrost', color='col', marker='o', alpha=0.5, data=dane)
plt.xlabel('Wiek')
plt.ylabel('Wzrost [cm]')
plt.title('Związek między wiekiem a wzrostem dzieci')
# Dopasowanie funkcji wielomianowej (liniowej) metodą najmniejszych kwadratów (regresja liniowa)
m, b = np.polyfit(dane['wiek'], dane['wzrost'], 1)
plt.plot(dane['wiek'], m * dane['wiek'] + b, color='black', linestyle='-.', linewidth=2)

# Zapisanie wykresu do pliku png i pdf
plt.savefig('przykład_1.2.png', dpi=300, bbox_inches='tight')
plt.savefig('przykład_1.2.pdf')

plt.show()
```

## 2. Matplotlib - obiektowy interfejs "Axes"

Poza podstawowym stylem używania biblioteki matplotlib przedstawionym wyżej, istnieje też styl obiektowy służący do tworzenia bardziej rozbudowanych wykresów oraz bezpośredniego operowania na poszczególnych elementach wykresu. Tu może warto wyjaśnić podstawowe pojęcia Figure i Axes. Figure to obiekt reprezentujący całkowitą powierzchnięc rysowania (może mieć wiele wykresów) oraz pojemnik zawierający wszystkie elementy rysunku. W jego obrębie są umieszczane właściwe wykresy reprezentowane przez obiekty klasy Axes (osie, panele). Każdy taki obiekt posiada informacje o pojedynczym wykresie: system współrzędnych, osie, etykiety, tytuł, jak i inne elementy wykresu. Poniżej przedstawiamy elementy rysunku w matplotlib (źródło: [https://matplotlib.org/](https://matplotlib.org/)), gdzie do każdego elementu rysunku odwołujemy się bezpośrednio:

Pyplot jest bardziej zwięzły, ale mniej elastyczny niż Axes. Jednak większośc funkcji pyplot ma swoje odpowiedniki w [matplotlib.axes.Axes](https://matplotlib.org/stable/api/_as_gen/matplotlib.axes.Axes.html), gdyż zwykle są one jedynie funkcjami opakowującymi. Przykładowo, odpowiednikiem pierwszego wykresu w tym notatniku używającym bezpośredniego odwołania do obiektu klasy Axes jest:

```python
# Dodaj obiekt klasy Axes (właściwy wykres) do aktualnego rysunku
fig, ax = plt.subplots()
# Narysuje wykres liniowy na osi ax
ax.plot(x, y)
# Wyświetl wykres
plt.show()
```

Inny przykład to heatmap wraz ze zmianą wielu parametrów wykresu korzystając z interfejsu Axes:

```python
import numpy as np

# Przygotuj dane ekspresji genów (wartości znormalizowane/log2 fold change)
np.random.seed(1024)
genes = ['BRCA1', 'TP53', 'EGFR', 'MYC', 'KRAS', 'PTEN', 'AKT1']
conditions = ['control', 'treatment A', 'treatment B', 'treatment C']

# Symuluj dane ekspresji (log2 fold change od -3 do 3)
expression = np.random.uniform(-2.5, 2.5, size=(len(genes), len(conditions)))

# Stwórz wykres
fig, ax = plt.subplots(figsize=(8, 6))
im = ax.imshow(expression, cmap='RdBu_r', aspect='auto', vmin=-3, vmax=3)

# Ustaw etykiety osi
ax.set_xticks(range(len(conditions)), labels=conditions,
              rotation=45, ha="right", rotation_mode="anchor")
ax.set_yticks(range(len(genes)), labels=genes)

# Dodaj wartości w komórkach
for i in range(len(genes)):
    for j in range(len(conditions)):
        text = ax.text(j, i, f'{expression[i, j]:.2f}', ha="center", va="center", color="white" if abs(expression[i, j]) > 1.5 else "black", fontsize=9)

# Dodaj colorbar i tytuł
cbar = fig.colorbar(im, ax=ax, label='Log2 Fold Change')
ax.set_title('Gene expression across conditions')
ax.set_xlabel('conditions')
ax.set_ylabel('genes')

plt.show()
```

Bardzo łatwo jest ustawić wspólne osie dla różnych paneli, co ułatwia porównania między panelami:

```python
# Wspólne osie przy porównywaniu ekspresji genów
fig, axes = plt.subplots(2, 1, figsize=(10, 5), sharex=True, sharey=True)

geny = ['BRCA1', 'TP53']
czas = np.arange(0, 24, 2)  # punkty czasowe (godziny)
np.random.seed(1024)

for i, (ax, gen) in enumerate(zip(axes, geny)):
    ekspresja = np.random.exponential(2, len(czas)) + i*2 # losowe dane z przesunięcciem
    ax.plot(czas, ekspresja, 'o-', linewidth=2, markersize=6)
    ax.set_ylabel(f'Ekspresja\n{gen}', fontsize=11)
    ax.grid(True, alpha=0.3)

axes[-1].set_xlabel('Czas (godziny)', fontsize=11)
fig.suptitle('Dynamika ekspresji genów w czasie', fontsize=14, fontweight='bold')
plt.show()
```

Teraz możemy tworzyć również bardziej zaawansowane wykresy. Poniżej przedstawiamy twójwymiarowy **wykres powierzchniowy** dla funkcji dwóch zmiennych (funkcja sinusoidalna). Tworzymy tu obiekt odpowiedniej podklasy klasy Axes (*projection*), która jest odpowiedzialna za realizację wykresów 3D, t.j. klasy mpl_toolkits.mplot3d.axes3d.Axes3D. Więcej rodzajów wykresów 3D można znaleźć w [dokumentacji](https://matplotlib.org/stable/api/toolkits/mplot3d/axes3d.html). Biblioteka matplotlib umożliwia również na tworzenie wykresów w innych układach współrzędnych poprzez odpowiednie rzutowania (projection) zdefiniowane jako podklasy klasy Axes.

```python
# Stwórz rysunek (Figure) i oś (obiekt podklasy klasy Axes) - właściwy wykres 3D
fig = plt.figure()
ax = fig.add_subplot(projection='3d')

# Przygotuj dane wejściowe
x_3d = np.linspace(-5, 5, 100)
y_3d = np.linspace(-5, 5, 100)
x_3d, y_3d = np.meshgrid(x_3d, y_3d)
z_3d = np.sin(np.sqrt(x_3d**2 + y_3d**2))

# Narysuj wykres powierzchniowy (używając mapy kolorów: viridis)
surf = ax.plot_surface(x_3d, y_3d, z_3d, cmap='viridis')

# Dodaj etykiety i tytuł
ax.set_xlabel('Oś X')
ax.set_ylabel('Oś Y')
ax.set_title('Przykładowy wykres Powierzchniowy 3D')

# Dodaj legendę (pasek kolorów), zmnijesz dwukrotnie, 
fig.colorbar(surf, shrink=0.5, aspect=10)

plt.show()
```

**A na koniec omawiania biblioteki matplotlib krótkie porównanie kiedy używać pyplot (plt.plot):**
- proste, jednorazowe wykresy,
- szybka eksploracja danych,
- interaktywna praca w Jupyter,

**a kiedy Axes API (fig, ax):**
- wiele podwykresów (subplots),
- precyzyjna kontrola nad elementami (np. dodatkowe i podwójne osie, niestandardowy podział paneli),
- publikacje naukowe,
- gdy chcesz zwrócić obiekt wykresu z funkcji.

[Plotly](https://plotly.com/) jest biblioteką do tworzenia interaktywnych wykresów i map dostępną dla różnych języków programowania (Python, R, Julia, Javascript, ggplo2, F#, MATLAB i Dash). Plotly można zainstalować za pomocą pip:

W plotly można tworzyć zarówno proste wykresy takie jak wykres liniowym, punktowy, słupkowy, itd, jaki i wykresy 3D, mapy i diagram, czy wykresy finansowe, bioinformatyczne, itd. Plotly posiada moduł [plotly.express](https://plotly.com/python/plotly-express/), który służy do tworzenia powszechnie używanych rodzajów wykresów w łatwy sposób. Domyśle wspiera on dane wejściowe w postacji pndas.DataFrame, ale inne typy danych takie jak tablice NumPy lub listy również są dopuszczalne. Nie będziemy tu omawiać dokładnie poszczególnych typów wykresów, ale podamy parę użytecznych przykładów.

Interaktywne funkcje tych wykresów w *plotly*:

- **hover**: najechanie myszką pokazuje szczegółowe wartości,
- **zoom**: przybliżanie i oddalanie (scroll lub box select),
- **pan**: przesuwanie wykresu (przeciągnij myszką),
- **export**: zapisywanie jako PNG (ikona aparatu),
- **autoscale**: automatyczne dopasowanie skali,
- **reset axes**: powrót do widoku początkowego.

**Przykład 3.1**: Wykres liniowy + zapisywanie do html (interaktywny wykres)

```python
import plotly.express as px
# Wczytaj dane gapminder
df = px.data.gapminder().query("continent == 'Europe'")
# Narysuj wykres liniowy, gdzie poszczególne kraje są oznaczone różnymi kolorami i znacznikami
fig = px.line(df, x='year', y='gdpPercap', color='country', symbol="country")

# Zapisz jako HTML (interaktywny)
fig.write_html("przykład_3.1.html")

fig.show()
```

**Przykład 3.2:** Interaktywna heatmapa ekspresji genów dla tych samych danych co poprzednia heatmapa

```python
# Tworzymy DataFrame dla łatwiejszej pracy
import pandas as pd
df_expression = pd.DataFrame(expression, index=genes, columns=conditions)

fig = px.imshow(df_expression,
                labels=dict(x="Warunki eksperymentalne", y="Geny", color="Log2 FC"),
                x=conditions,
                y=genes,
                color_continuous_scale='RdBu_r',  # Czerwony-Biały-Niebieski (odwrócony)
                color_continuous_midpoint=0,  # Wyśrodkuj na 0
                aspect='auto',
                title='Heatmapa ekspresji genów - interaktywna')

# Dostosuj wygląd
fig.update_xaxes(side="bottom")
fig.update_layout(
    width=700,
    height=500,
    font=dict(size=12)
)

fig.show()
```

Czasami plotly.express nie jest w stanie wygenerować wykresu, który chcemy ze względu na drobne szczegóły lub brak odpowiedniej funkcji. Wtedy możemy przejść do modułu *plotly.graph_objects* dostarczającego więcej rodzajów wykresów i sposobów ich dostosowania.

**Przykład 3.5**: Wykres liniowy w *plotly.graph_objects*

```python
import plotly.graph_objects as go
import plotly.express as px

# Przykładowe dane
import numpy as np
np.random.seed(1)
random_x = np.linspace(0, 1, 100)
random_y = np.random.randn(100)

# Stwórz rysunek i kolejne jego elementy (wykresy liniowe przy użyciu go.Scatter) używając różnych stylów
fig = go.Figure()
fig.add_trace(go.Scatter(x=random_x, y=random_y-5, mode='lines', name='lines'))
fig.add_trace(go.Scatter(x=random_x, y=random_y  , mode='lines+markers', name='lines+markers'))
fig.add_trace(go.Scatter(x=random_x, y=random_y+5, mode='markers', name='markers'))
fig.show()
```

Plotly natomiast oferuje większą estetykę i łatwość integracji z frameworkiem Dash, co czyni go doskonałym wyborem do tworzenia interaktywnych raportów.

**Przykład 4.1**: Podstawowy scatter plot dla klasycznego zbioru *iris*. Bokeh automatycznie dodaje interaktywne narzędzia.

```python
from bokeh.plotting import figure, show, output_notebook

# Konfiguracja dla Jupyter Notebook
# wpp wygeneruje się tymczasowy html i zostanie otworzony w przeglądarce
output_notebook()
# Przełączenie na output do pliku
# output_file('przykład_4.1.html')

iris = pd.read_csv('iris.csv')

# Tworzenie wykresu
p = figure(
    width=700, 
    height=500,
    title="Długość vs szerokość płatka (iris)",
    x_axis_label='Długość płatka (cm)',
    y_axis_label='Szerokość płatka (cm)'
)

# Dodanie punktów
p.circle(
    x=iris['petal.length'],
    y=iris['petal.width'],
    size=10,
    color='navy',
    alpha=0.5
)

show(p)
```

**Przykład 4.2**: Boxplot

Bokeh nie ma niektórych standardowych funkcji, np. histogram (możemy go stworzyć używając `quad` (prostokąty)) czy boxplot (wymaga ręcznego obliczenia statystyk i rysowania każdego elementu wykresu).


```python
# Obliczanie histogramu za pomocą numpy
hist, edges = np.histogram(iris['petal.length'], bins=20)

p = figure(
    width=800,
    height=500,
    title="Histogram długości płatków",
    x_axis_label='Długość płatka (cm)',
    y_axis_label='Liczba obserwacji'
)

# Rysowanie prostokątów
p.quad(
    top=hist,
    bottom=0,
    left=edges[:-1],
    right=edges[1:],
    fill_color='navy',
    line_color='white',
    alpha=0.7
)

show(p)
```

**Przykład 4.3**: Linked plots - połączone wykresy

Jedną z ciekawszych funkcji *Bokeh* jest możliwość połączenia wykresów, np. gdy zaznaczysz punkty na jednym wykresie, zostaną one automatycznie zaznaczone na drugim, ale również wiele innych możliwości interakcji  między różnymi panelami (wykresami). Tu mamy też przykład łączenia wykresów w jeden układ oraz definiowania tooltip'ów.

```python
from bokeh.models import HoverTool, ColumnDataSource
from bokeh.transform import factor_cmap
from bokeh.palettes import Category20_3
from bokeh.layouts import row

# Wspólne źródło danych dla obu wykresów
# Bokeh nie pozwala na "." w nazwach kolumn przy "hover"
iris2 = iris.copy()
iris2.columns = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width', 'variety']
source = ColumnDataSource(iris2)

# Definicja kolorów (mapowanie)
color_mapper = factor_cmap(
    field_name='variety',
    palette=Category20_3,
    factors=iris['variety'].unique().tolist()
)

# Definicja tooltipów
hover = HoverTool(tooltips=[
    ('Indeks', '@index'),
    ('Gatunek', '@variety'),
    ('Długość płatka', '@petal_length{0.0} cm'),
    ('Szerokość płatka', '@petal_width{0.0} cm'),
    ('Długość działki', '@sepal_length{0.0} cm'),
    ('Szerokość działki', '@sepal_width{0.0} cm')
])
# Wykres 1: petal_length vs petal_width
left = figure(
    width=500,
    height=500,
    title="Płatki (kliknij i przeciągnij aby zaznaczyć)",
    x_axis_label='Długość płatka (cm)',
    y_axis_label='Szerokość płatka (cm)',
    tools=[hover, 'box_select', 'lasso_select', 'reset', 'save']
    #tools='box_select,lasso_select,reset'
)

left.circle(
    x='petal_length',
    y='petal_width',
    source=source,
    size=10,
    color=color_mapper,
    alpha=0.6,
    selection_color='red',
    nonselection_alpha=0.1
)

# Wykres 2: sepal_length vs sepal_width
right = figure(
    width=500,
    height=500,
    title="Działki kielicha (zsynchronizowane zaznaczenie)",
    x_axis_label='Długość działki (cm)',
    y_axis_label='Szerokość działki (cm)',
    tools=[hover, 'box_select', 'lasso_select', 'reset', 'save']
)

right.circle(
    x='sepal_length',
    y='sepal_width',
    source=source,  # To samo źródło danych!
    size=10,
    color=color_mapper,
    alpha=0.6,
    selection_color='red',
    nonselection_alpha=0.1
)

# Układamy wykresy obok siebie
layout = row(left, right)

show(layout)
```

**Przykład 5.3**: *Jointplot* tworzy wykres pokazujący relację między dwiema zmiennymi (w centrum) wraz z ich rozkładami brzegowymi (marginalnymi) na osiach

Ciekawe opcje:

- *kind='kde', fill=True* - tworzy wypełniony wykres gęstości, gdzie intensywność koloru wskazuje na koncentrację danych,
- *hue='variety'* - koloruje punkty według wartości w kolumnie 'variety' (podział na grupy).
