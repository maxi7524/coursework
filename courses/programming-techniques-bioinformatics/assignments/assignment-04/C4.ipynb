## Ćwiczenie 1

Stwórz histogram pokazujący rozkład długości płatków (petal.length) w zbiorze `iris`.

Wymagania:
- Użyj matplotlib.pyplot
- Ustaw 20 bins (przedziałów)
- Dodaj tytuł, etykiety osi x i y
- Dodaj pionową linię pokazującą średnią wartość (użyj `plt.axvline`)
- Dodaj legendę

(*)Wersja rozszerzona dla chętnych: 

- Stwórz drugi wykres pokazujący histogram z podziałem na gatunki (`plt.hist` z parametrem `label`). 
- Spróbuj również z użyciem biblioteki *seaborn* (`sns.histplot` z parametrem `kde=True` (Kernel Density Estimation))

```python
iris = pd.read_csv('../iris.csv')
iris['variety']
```

```python
# bardzo na szybko
iris['petal.length'].hist()
plt.axvline(iris['petal.length'].mean(), color='Black')
```

```python
# histogram z podziałem na gatunki
plt.hist(iris['petal.length'],
    bins=20, density=True)
plt.title('Iris histogram')
plt.xlabel('Długość płatka')
plt.ylabel('empiryczna gęstość')
plt.axvline(iris['petal.length'].mean(), color='Black')
plt.legend()
```

```python
# histogram z podziałem na gatunki
plt.hist(iris['petal.length'],
    bins=20, density=True, label=iris['variety'])
plt.title('Iris histogram')
plt.xlabel('Długość płatka')
plt.ylabel('empiryczna gęstość')
plt.axvline(iris['petal.length'].mean(), color='Black')
plt.legend()
```

```python
feature = 'petal.length'
for variety, subset in iris.groupby('variety'):
    plt.hist(subset[feature],
             bins=10,
             density=True, # 
             label=variety)
plt.legend()
```

## Ćwiczenie 4
Stwórz interaktywny wykres 3D pokazujący trzy wymiary danych `iris` jednocześnie (`sepal.length`, `sepal.width`, `petal.length`).

Wymagania:

- Użyj `plotly.express.scatter_3d`
- Pokoloruj punkty według gatunku
- Ustaw rozmiar punktów proporcjonalnie do `petal.width`
- Dodaj etykiety osi
- Dodaj tytuł
- Bonus: zapisz do pliku html (`write_html`)
