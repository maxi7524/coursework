```python
# def mcmc_prymitive_sample(n: int, dist: stats.rv_frozen, ):

# params
n_experiment = 1_000
n_probe = 1_000_000


def raw_mcmc(n_experiment, n_probe, dist, condition):
    '''condition is giving us set $(condition, \infty)$ '''
    fractions = []
    for _ in range(n_experiment):
        probe = dist.rvs(n_probe)
        fraction = (probe > condition).sum() / n_probe
        fractions.append(fraction)

    return np.mean(fractions), 2 * np.std(fractions) / np.sqrt(n_experiment)

raw_mcmc(n_experiment=n_experiment, n_probe=n_probe, dist=stats.norm(0, 1), condition=4)
```

```python
n_experiment = 1_000
n_probe = 1_000

def importance_sampling_mcmc(n_experiment, n_probe, dist_p, dist_q, condition):
    '''
    Here we move distribution by condition (so it is in right area)
    '''
    fractions = []
    for _ in range(n_experiment):
        probe = dist_q.rvs(n_probe) + condition
        weight = lambda x: dist_p.pdf(x) / (dist_q.pdf(x - condition))
        fraction = weight(probe) * (probe > condition).sum() / n_probe
        fractions.append( fraction)

    return np.mean(fractions), 2 * np.std(fractions) / np.sqrt(n_experiment)
    
importance_sampling_mcmc(n_experiment=n_experiment, n_probe=n_probe, dist_p=stats.norm(0, 1), dist_q=stats.expon(1), condition=4)
```
