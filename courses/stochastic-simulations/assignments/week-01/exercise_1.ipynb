## Exercise 1

### 1. Make following distribution 

$$X = \sum_{i=1}^{12} U_i -6$$

To do it implement:
- random variable generator 
- pdf of $X$

### 2. Calculate statistics
Calculate mean and variance

### 3. Plots

Plot histogram (with density setting) and pdf (we can assume normal) of our distribution

### 4. Normality test

Apply normality test 

```python
# produce X as sum 12 uniform ASSUMPTION - we use pseudo random numbers - we skip sp 

## parameters
N_probe = 1000

## Variable
### Generating random variable
X_rvs = lambda n:  np.sum([stats.uniform(0, 1).rvs(n) for _ in range(12)], axis=0) - 6 
### Generating pdf of variable 
# X_pdf = lambda x: np.sum([stats.uniform(0, 1).pdf(x) for _ in range(12)], axis=0) - 6 
### here we obtain (12, n) sum over rows (axis = 0) and use broadcasting
## chcemy dostać n prób losowych 
probe = X_rvs(N_probe)
```

```python
probe.mean(), probe.var()
```

```python
## 3. 
### Parameters
X_domain = np.linspace(-3, 3, 400)
### Plot histogram
sns.histplot(probe, stat='density')
### TODO plot pdf 
plt.plot(X_domain, stats.distributions.norm().pdf(X_domain), color='red')
```

```python
## 4. Testing normality
stats.shapiro(probe) 
```

## Exercise 1.8 p 27

### Book part
(Arcsine Law). Let $X_1, \dots, X_n$ be independent and identically distributed (i.i.d.) random variables with expected value $\mathbb{E}X_i = 0$ and finite variance. Let $S_i = X_1 + \dots + X_i$ for $i = 1, \dots, n$. Define

$$T_n = \frac{1}{n} \sum_{i=1}^{n} \mathbb{1}(S_i > 0).$$

Interpret $T_n$ in terms of a "sequence of games" with payoffs $X_i$. Plot the trajectory $(0, S_1, S_2, \dots, S_n)$ and calculate the corresponding value of $T_n$. Experimentally investigate the probability distribution of the random variable $T_n$ for large $n$. Verify experimentally that for $n \to \infty$, $T_n \to \text{Beta}(1/2, 1/2)$ (the name of the theorem comes from the cumulative distribution function of this beta distribution).

### Parameters

We assume $X_i \sim \mathcal{N}(0, 1)$

```python
# parameters
N_simulation = 1_000_000
N_probe = 1000
# code
norm = stats.distributions.norm()


T_list = list()
for _ in range(N_simulation):
    probe = np.array(norm.rvs(N_probe))
    T = (probe.cumsum() > 0).sum() / N_probe
    T_list.append(T)
```

```python
X_domain = np.linspace(0, 1, 1000)
sns.histplot(T_list, stat='density') # to jest rozkład wszystkich 
plt.plot(X_domain, stats.distributions.beta(1/2, 1/2).pdf(X_domain), color='red')
```
