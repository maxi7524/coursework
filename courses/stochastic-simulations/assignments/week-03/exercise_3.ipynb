#### Theoretical results
##### cdf
We want to find conditional cdf (It depends only on recent value), we write that as:
$$\begin{align*}

F_{R_{i}|R_{i-1}}(y|x) & =cdf_{R_i|R_{i-1}}(y|x)  = \mathbb{P}[R_{i} \leq y|R_{i-1}=x] = \mathbb{P}[X\leq y|X>x]= \frac{\mathbb{P}[X\leq y, X>x]}{1-\mathbb{P}[X\leq x]}=  \\

 & = \frac{{\mathrm{cdf}_{X}(y) - \mathrm{cdf}_{X}(x)}}{1-\mathrm{cdf}_{X}(x)}

\end{align*}$$
##### pdf 
To obtain pdf de differentiate function due to $y$:
$$\begin{align*}
f_{R_{i}|R_{i-1}}(y|x) &=\mathrm{pdf}_{R_{i}|R_{i-1}}(y|x) & =\frac{\mathrm{d}}{\mathrm{d}y}\mathrm{cdf_{R_{i}|R_{i-1}}(y|x)} = \\ &= \frac{{\mathrm{pdf}_{X}(y)}}{1-\mathrm{cdf}_{X}(x)},  & y & >x
\end{align*}$$
##### marginal distributions
To obtain marginal distribution we obtain from definition
$$f_{R_{i}}(y)=\int_{-\infty}^{y}f_{R_{i}|R_{i-1}}(y|x)\underbrace {f_{R_{i-1}}(x)}_{\text{recurrence}}\mathrm{d}x$$
But here we obtain recurennce relation. 
##### Recurrence
Marginal for $R_1$": 
$$\begin{align}
f_{R_{1}}(y) & =\int_{-\infty}^{y}f_{R_{1}|R_{0}}(y|x){f_{R_{0}}(x)}\mathrm{d}x = \int_{-\infty}^{y} \frac{{f_{X}(y)}}{1-F_{X}(x)}f_{R_{0}=X}(x)\mathrm{d}x= \\
	& = \left[\begin{cases} 
u=F_{X}(x) \\
\mathrm{d}u=f_{X}(x)\mathrm{d}x
\end{cases}\right] = f_{X}(y)\int_{0}^{F_{X}(y)} \frac{1}{1-u}\mathrm{d}u = f_{X}(y)[-\ln(1-u)|_{0}^{F_{X}(y)}] = \\
 & = f_{X}(y)(-\ln(1-F_{X}(y))) 
\end{align}$$
Marginal for $R_2$"
$$\begin{align}
f_{R_{2}}(y) & =\int_{-\infty}^{y}f_{R_{2}|R_{1}}(y|x){f_{R_{1}}(x)}\mathrm{d}x = \int_{-\infty}^{y} \frac{{f_{X}(y)}}{1-F_{X}(x)}f_{R_{1}}(x)\mathrm{d}x=  \\

u=F_{X}(x) \\

 & =\frac{f_{X}(y)}{n!} [-\ln(1-F_{X}(y))]^{n}
\end{align}$$

#### Simulation solution
To simulate process efficiently we use inverse sampling to always obtain suitable sample:
1. We generate $\mathrm{probe} \sim F$ at random 
2. We take $l_i = \mathrm{cdf}_F(\mathrm{probe}_{i-1})$
3. We take $X_i \sim \mathrm{Unif}([l, 1])$
4. We take $R_i = F^{-1}(X)$
5. repeat 2-4

```python
# #TODO 
def rvs_record_single(k: int, dist: stats.rv_continuous | stats.rv_discrete):
    '''
    In this solution i use different approach when i will be able to accept all probes using inverse sampling method:
    - Generate first probe
    - Take it through distribution and divide domain X ~ U(dist.cdf(probe(i)), 1)
    - we than take dist.ppf(X) (inverse of cdf) 
    '''
    # params 
    ## here we take n+1 probes we omit first one 
    records = np.zeros(k+1)
    curr_max = 0 
    curr_size = 0
    ## first record
    records[0] = dist.rvs()
    
    # obtaining other index
    for i in range(1, k+1):
        # cdf (probe[i-1])
        left_margin = dist.cdf(records[i-1])
        # X ~ Unif(cdf, 1)
        sample = stats.uniform(left_margin, 1-left_margin).rvs()
        # cdf^-1 (X)
        records[i] = dist.ppf(sample) 
           
    return records

    ## helper function

def rvs_record(n: int, k: int, dist: stats.rv_continuous | stats.rv_discrete):
    '''
    Wrapper for taking n samples from this distribution
    '''
    # generate multiple probes    
    probes = np.array([rvs_record_single(k, dist) for _ in range(n)])
    # reshape it into (n x k) array
    return probes

def plot_theoretical(k: int, domain: list[int],dist: stats.rv_discrete | stats.rv_continuous):
    # unpack border
    l, r = domain
    X_dom = np.linspace(l, r, 100 * (r-l))

    numerator  = dist.pdf(X_dom) * (1 - np.log(1-dist.cdf(X_dom)))**k
    denominator = math.factorial(k)
    return X_dom, numerator/denominator
```

```python
# params
size = 10_000
k = 5

# test definitions
tests = {
    'Uniform': [None, [0, 1], stats.uniform(0, 1)], 
    'Pareto': [None, [0, 10], stats.pareto(b=1)],
    'Exponential': [None, [0, 10], stats.expon(scale= 1/2)]
}

for test, params in tests.items():
    tests[test][0] = rvs_record(size, k, params[2])
```

```python

# plots
## distributions for 1-6
for key, [probe, domain, dist] in tests.items():
    for i in range(1, 6):
        # plot histogram
        sns.histplot(probe[:, i], stat='density', label=f'$R_{i}$')
        # plot density 
        X_dom, values = plot_theoretical(i, domain, dist)
        plt.plot(X_dom, values, label=f'Theoretical: $R_{i}$')

    plt.legend()
    plt.title(f'Pdf and histograms of {key}')
    plt.show()
```
