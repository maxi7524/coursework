### Markov chains 

Our task is to simulate markov chains. We want to obtain trajectories. 

It is defined as 
- $1 ->1$ with $1=\alpha$ 
- $1 ->2$ with $\alpha$ 
- $2->2$ with $1-\beta$
- $2->1$ woth $\beta$

#### Obtain distributions


#### Obtain limit


```python
def markov_rvs(alpha: float, beta: float, n: int, L:int, init_prob: list[float, float]):
    '''
    Simulation of Random Markov process (two states)
    '''
    
    # params
    ## rtn list
    rtn_val = np.zeros([n, L])
    
    # simulation 
    ## init probabilities
    rtn_val[:, 0] = np.random.choice([1, 2], n, p=init_prob)
    
    ## 1:n steps
    for t in range(L):
        ## identify states with ones
        mask_state_one = rtn_val[:, t-1] == 1
        k = np.sum(mask_state_one)
        
        rtn_val[mask_state_one, t] = np.random.choice([1, 2], k, p=[1-alpha, alpha])

        rtn_val[~mask_state_one, t] = np.random.choice([1, 2], n-k, p=[beta, 1-beta])   

    return rtn_val

```

```python
probe = markov_rvs(0.1, 0.2, 1, 1000, [0.5, 0.5])
probe[probe==1].sum() / 1000 
```

```python
n = 1
L = 1_000_000

probe = markov_rvs(0.9, 0.8, n, L, [0.5, 0.5])
(probe == 1).mean()
```
