```python
import numpy as np
from utils_algorithms_PF_and_FFBS import (
    generate_continuous_hmm_data,
    particle_filter_continuous,
)

import matplotlib.pyplot as plt

# Your model parameters
alpha_value = 1.05   # Example autoregression parameter
sim_length = 20 
num_particles = 10**4
sigma_x = 1.0  # Variance = 1 -> std = 1
sigma_y = 3.0  # Variance = 9 -> std = 3

# 1. Continuous data generation
true_states, observations = generate_continuous_hmm_data(
    alpha_param=alpha_value, sim_length=sim_length, sigma_x=sigma_x, sigma_y=sigma_y
)

# 2. Running Particle Filter (with resampling)
_, pf_means, pf_stds = particle_filter_continuous(
    alpha_param=alpha_value, observations=observations, num_particles=num_particles, 
    sigma_x=sigma_x, sigma_y=sigma_y, use_resampling=True
)

# 3. Running SIS (without resampling for comparison)
_, sis_means, sis_stds = particle_filter_continuous(
    alpha_param=alpha_value, observations=observations, num_particles=num_particles, 
    sigma_x=sigma_x, sigma_y=sigma_y, use_resampling=False
)

# 4. Continuous state tracking plot
plt.figure(figsize=(15, 7))
time_axis = np.arange(sim_length)

# Plotting Ground Truth and Observations
plt.plot(time_axis, true_states, color='black', linewidth=2, label='True Hidden State (Ground Truth)')
plt.scatter(time_axis, observations, color='gray', alpha=0.5, s=15, label='Noisy Observations (Y_t)')

# Plotting estimate from Particle Filter (Mean + 2*sigma band)
plt.plot(time_axis, pf_means, color='teal', linestyle='-', label='PF Estimated Mean')
plt.fill_between(time_axis, pf_means - 2*pf_stds, pf_means + 2*pf_stds, 
                 color='teal', alpha=0.2, label='PF 95% Confidence Interval')

plt.title('Continuous State Estimation using Particle Filter')
plt.xlabel('Time (t)')
plt.ylabel('State Value (X_t)')
plt.legend(loc='upper left')
plt.grid(True, linestyle=':')
plt.show()
```
